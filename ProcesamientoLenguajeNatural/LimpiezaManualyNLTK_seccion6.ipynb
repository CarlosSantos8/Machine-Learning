* **Tokenización por Espacios en Blanco:** El texto se dividió usando un `.split()` simple. Esto provocó que los signos de puntuación se quedaran adheridos a las palabras (ej. `'vermin.'`, `'back,'`), generando tokens duplicados para el modelo.
* **Problema de Capitalización:** Coexisten palabras con mayúscula inicial y minúscula (ej. `'He'` y `'he'`). Para el vocabulario del modelo, representarán dos vectores o IDs completamente distintos, fragmentando el aprendizaje.
* **Ruido en Contracciones y Citas:** Tokens como `'"What\'s'` o `'me?"'` arrastran comillas y signos de interrogación pegados, lo que ensuciará la matriz de embeddings.

# <font color="#004D7F" size=5>3.3. Seleccionar palabras</font>

Otro enfoque podría ser usar el modelo `regex` de expresiones regulares (`re`) y dividir el documento en palabras seleccionando cadenas de caracteres alfanuméricos (az, AZ, 0-9 y ' ').

`````python
import re

# Importamos el módulo de expresiones regulares para manipulación avanzada de texto.
import re

# Dividimos el texto usando como separador cualquier carácter que NO sea alfanumérico (\W).
# El símbolo '+' agrupa caracteres consecutivos (como puntos, comas o espacios juntos).
# Nota técnica: Esto eliminará la puntuación, pero separará contracciones (don't -> don, t) 
# y palabras compuestas con guion (armour-like -> armour, like).
words = re.split(r'\W+', text)

# Mostramos los primeros 100 tokens para validar visualmente el resultado del split
# sin saturar la pantalla de Colab con todo el dataset.
print(words[:100])
`````

# <font color="#004D7F" size=5>3.4. Dividir por espacios en blanco y quitar puntuación</font>

Es posible que queramos las palabras, pero sin la puntuación como las comas y las comillas. También queremos mantener las contracciones juntas. 

Python proporciona una constante llamada `string.punctuation` que proporciona una excelente lista de caracteres de puntuación. Por ejemplo:

`````python
import string

# Importamos el módulo nativo 'string' de Python para acceder a constantes de texto predefinidas.
import string

# Imprime una cadena que contiene todos los caracteres considerados como puntuación en el sistema ASCII.
# El output será: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
# Nota técnica: Es una herramienta útil para limpieza rápida (por ejemplo, pasándola a un método .strip() 
# o un diccionario de traducción), pero ten en cuenta que no incluye caracteres especiales de otros 
# idiomas (como los signos de interrogación '¿' o exclamación '¡' invertidos del español).
print(string.punctuation)
`````

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
Más información sobre [`string.punctuation`](https://docs.python.org/3/library/string.html?highlight=string%20punctuation#string.punctuation)

Podemos usar expresiones regulares para seleccionar los caracteres de puntuación y usar la función `sub()` para reemplazarlos con nada. 

`````python
import re
import string

# 1. Separación inicial por espacios en blanco.
# Genera una lista de tokens donde la puntuación sigue adherida a los extremos de las palabras (ej. 'vermin.').
words = text.split()

# 2. Compilación de la expresión regular para optimizar el rendimiento.
# - re.escape(string.punctuation) escapa de forma segura los caracteres especiales de puntuación de Python.
# - '[%s]' define una clase de caracteres (character class). Cualquier carácter que coincida con la lista será seleccionado.
# - re.compile acelera el procesamiento si esta regla se va a aplicar de forma iterativa sobre miles de tokens.
re_punc = re.compile('[%s]' % re.escape(string.punctuation))

# 3. Limpieza de puntuación por token mediante List Comprehension.
# Se itera sobre cada palabra 'w' de la lista 'words' y se reemplaza (.sub) cualquier carácter de puntuación por un string vacío ''.
# Nota técnica: A diferencia de re.split(r'\W+'), este enfoque mantiene unidas las palabras compuestas y contracciones,
# eliminando la puntuación de sus extremos o colapsándola (ej. 'armour-like' -> 'armourlike', "What's" -> 'Whats').
stripped = [re_punc.sub('', w) for w in words]

# 4. Inspección y comparación de resultados con formato claro.
# Se utilizan saltos de línea (\n) y separadores visuales para diferenciar claramente 
# el estado de los tokens antes y después del proceso de limpieza en el entorno de Colab.
print("=== TOKENS ORIGINALES (CON PUNTUACIÓN) ===")
print(words[:100])
print("\n" + "="*50 + "\n")
print("=== TOKENS LIMPIOS (SIN PUNTUACIÓN) ===")
print(stripped[:100])
`````

# <font color="#004D7F" size=5>3.5. Caso de normalización</font>

Es común convertir todas las palabras a un caso. 

Podemos convertir todas las palabras a minúsculas llamando a la función `lower()` en cada palabra. 

`````python
# 1. Separación inicial por espacios en blanco.
# Divide el string 'text' utilizando cualquier espacio, tabulación o salto de línea como delimitador.
words = text.split()

# 2. Normalización de caja (Lowercasing) mediante List Comprehension.
# Transforma cada token de la lista a minúsculas. Esto unifica palabras idénticas con distinta capitalización
# (ej. 'He' y 'he' ahora se registrarán bajo el mismo ID o vector), consolidando el tamaño del vocabulario.
words = [word.lower() for word in words]

# 3. Impresión y monitoreo del pipeline en Colab.
# Muestra los primeros 100 elementos resultantes para validar que la conversión a minúsculas 
# se haya ejecutado correctamente sobre los tokens que aún conservan la puntuación adjunta.
print("=== TOKENS ORIGINALES (CON PUNTUACIÓN) ===")
print(words[:100])
print("\n" + "="*50 + "\n")
`````

Poder utilizar la función `sent_tokenize` debe de estar previametne descargado

`````python
import nltk

# 1. Importación de la librería NLTK (Natural Language Toolkit).
# Herramienta fundamental para el procesamiento de lenguaje natural tradicional.
import nltk

# 2. Descarga directa y no interactiva del paquete 'punkt'.
# 'punkt' contiene los modelos de tokenización basados en aprendizaje no supervisado (Punkt Sentence Tokenizer).
# Este modelo ha sido entrenado para entender dónde termina una oración y dónde empieza una palabra, 
# reconociendo correctamente abreviaturas (como "Mr.", "Dr.") y signos de puntuación sin romper el texto de forma ingenua.
# Nota técnica: Al especificar 'punkt' entre comillas, la descarga se realiza de forma silenciosa y automática,
# evitando que la celda de Colab se congele esperando una interacción manual del usuario.
nltk.download('punkt')
`````

# <font color="#004D7F" size=5>4.3. Dividir en palabras</font>

NLTK proporciona una función llamada `word_tokenize()` para dividir cadenas en tokens (nominalmente palabras) que divide tokens en función de los espacios en blanco y la puntuación. 

`````python
from nltk.tokenize import word_tokenize

# 1. Importación del tokenizador de palabras de NLTK.
# word_tokenize es un envoltorio (wrapper) que utiliza internamente una adaptación 
# del algoritmo Treebank Word Tokenizer junto con el modelo 'punkt'.
from nltk.tokenize import word_tokenize 

# 2. Tokenización avanzada a nivel de palabra.
# A diferencia de un split() básico por espacios, este método separa de forma inteligente 
# los signos de puntuación de las palabras, convirtiéndolos en tokens individuales (ej. 'vermin', '.').
# También maneja de forma estandarizada las contracciones en inglés (ej. "don't" se divide en ['do', "n't"]).
tokens = word_tokenize(text)

# 3. Inspección visual de la muestra en la celda de Colab.
# Se extraen los primeros 100 tokens para evaluar la estructura sintáctica generada 
# y verificar si se conserva la integridad semántica de los términos clave.
print(tokens[:100])
`````

- **Tokenización de Puntuación Perfecta:** A diferencia del .split() inicial, los signos de puntuación como las comas (',') y los puntos ('.') ya no están pegados a las palabras; ahora son tokens independientes en la lista. Esto preserva la estructura gramatical para el análisis secuencial.

- **Manejo Exitoso de Palabras Compuestas:** El término de ingeniería lingüística 'armour-like' se mantuvo unificado en un solo token. Esto es excelente para no perder la semántica del adjetivo compuesto.

- **Separación de Contracciones Estilo Penn Treebank:** La contracción "What's" fue dividida en dos tokens distintos: ['What', "'s"]. Para tareas de NLP (como el etiquetado sintáctico), esto es la norma técnica correcta, ya que separa el sujeto (What) del verbo/morfema ('s).

- **Artefacto de Comillas del Modelo de NLTK:** El token '```` representa la comilla doble de apertura antes de *"What's"*. `word_tokenize` de NLTK convierte las comillas tipográficas en el formato de codificación del Penn Treebank (dos acentos graves   ), lo cual es un comportamiento estándar pero genera un caracter que requerirá atención si no deseas ruido gráfico.

- **Problema de Capitalización Persistente:** Sigues teniendo tokens con mayúscula y minúscula coexistiendo (ej. 'One', 'Gregor', 'Samsa', 'He', 'His'). Mientras que para los nombres propios como Gregor Samsa es útil mantener las mayúsculas (para Reconocimiento de Entidades Nombradas o NER), para el resto de palabras generará duplicados semánticos.

# <font color="#004D7F" size=5>4.4. Filtrar la puntuación</font>

Podemos filtrar todos los tokens que no nos interesen, como todos los signos de puntuación independientes con la función `isalpha()`. Por ejemplo:

`````python
import nltk
from nltk.tokenize import word_tokenize

# 1. Tokenización avanzada inicial.
# Convierte el string 'text' en una lista de tokens, aislando la puntuación 
# y manejando adecuadamente las contracciones de forma nativa.
tokens = word_tokenize(text)

# 2. Filtrado de tokens no alfabéticos mediante List Comprehension.
# El método .isalpha() de Python evalúa cada token y solo conserva aquellos compuestos 
# estrictamente por caracteres del abecedario (A-Z, a-z).
# Nota técnica: Esto elimina de forma radical e inmediata todos los signos de puntuación aislados (',', '.'),
# los artefactos de comillas de NLTK ('``') y cualquier número presente.
#⚠️ ADVERTENCIA: Este filtro destruirá palabras compuestas con guion (ej. 'armour-like' se elimina por completo 
# de la lista al contener el caracter '-') y alterará contracciones (ej. "'s" se conserva porque contiene letras, 
# pero perderá su relación sintáctica original).
words = [word for word in tokens if word.isalpha()]

# 3. Monitoreo visual de la muestra en Colab.
# Imprime los primeros 100 elementos de la nueva lista puramente alfabética para validar el resultado.
print(words[:100])
`````

# <font color="#004D7F" size=5>4.5. Filtrar palabras vacías (y pipeline)</font>

Las palabras vacías son aquellas palabras que no contribuyen al significado más profundo de la frase que podemos quitar con `stopwords`.

`````python
import nltk
from nltk.corpus import stopwords

# 1. Importación del corpus de palabras vacías (Stopwords).
# NLTK provee listas predefinidas de palabras funcionales de alta frecuencia para múltiples idiomas.
from nltk.corpus import stopwords

# 2. Carga de la lista de stopwords específicas para el idioma inglés.
# Genera un set o lista con palabras como 'i', 'me', 'my', 'myself', 'you', 'your', etc.
# Nota técnica: Para que esta celda funcione sin errores en Colab, asegúrate de haber ejecutado 
# previamente el comando 'nltk.download('stopwords')'. De lo contrario, lanzará un LookupError.
stop_words = stopwords.words('english')

# 3. Monitoreo visual de la muestra en Colab.
# Imprime los primeros 10 elementos de la colección para verificar la correcta carga y estructura de los datos.
print(stop_words[:100])
`````

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i>
Más información sobre [`stopwords`](https://pythonprogramming.net/stop-words-nltk-tutorial/)

Vemos una recopilación que normalmente se usa en la limpieza:
- Cargar el texto sin procesar.
- Dividir en tokens.
- Convertir a minúsculas.
- Eliminar la puntuación de cada token.
- Filtrar los tokens restantes que no sean alfabéticos.
- Filtrar los tokens que son palabras vacías.

`````python
import string
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords


# split into words
# Utilizamos word_tokenize de NLTK para separar adecuadamente palabras,
# contracciones y signos de puntuación de forma inteligente.
tokens = word_tokenize(text)

# convert to lower case
# Normalizamos el texto pasando todos los tokens a minúsculas para consolidar el vocabulario.
tokens_lower = [word.lower() for word in tokens]

# prepare regex for char filtering
# Compilamos una expresión regular con todos los caracteres de puntuación estándar de Python.
re_punc = re.compile('[%s]' % re.escape(string.punctuation))

# remove punctuation from each word
# Sustituimos cualquier carácter de puntuación dentro de los tokens por un string vacío.
# Nota: Esto limpia signos en los extremos y colapsa puntuaciones internas.
stripped = [re_punc.sub('', w) for w in tokens_lower]

# remove remaining tokens that are not alphabetic
# Filtramos la lista para mantener únicamente los tokens compuestos por letras,
# lo que elimina eficazmente tokens vacíos ('' resultantes de la puntuación aislada) y números.
words = [word for word in stripped if word.isalpha()]

# filter out stop words
# Cargamos el conjunto de stopwords en inglés de NLTK y eliminamos las palabras funcionales
# de alta frecuencia para quedarnos únicamente con los tokens con mayor carga semántica.
stop_words = set(stopwords.words('english'))
words_final = [w for w in words if w not in stop_words]

# Imprimimos una muestra de control en Colab para verificar el pipeline completo
print(words_final[:100])
`````

Stemming se refiere al proceso de reducir cada palabra a su raíz o base. Por ejemplo, _fishing, fished, fisher_ todo se reduce a _fish_.

Está disponible en NLTK a través de la clase `PorterStemmer`. Por ejemplo:

`````python
# 1. Instanciación del algoritmo Porter Stemmer
# Inicializamos el objeto para aplicar la reducción morfológica basada en reglas heurísticas para inglés.
porter = PorterStemmer()

# 2. Aplicación de Stemming sobre la lista final limpia (words_final)
# Reducimos los tokens a su raíz morfológica o base (stem) de forma normalizada (ej. 'troubled' -> 'troubl').
# Al hacerlo después del filtrado de stopwords, evitamos procesar tokens innecesarios optimizando el pipeline.
words_stemmed = [porter.stem(word) for word in words_final]

# Imprimimos una muestra de control en Colab para verificar el pipeline completo con Stemming
print("=== RESULTADO FINAL CON STEMMING ===")
print(words_stemmed[:100])
`````
