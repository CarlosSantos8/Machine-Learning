# Inferencia Estadística

La **inferencia estadística** es el proceso mediante el cual **hacemos afirmaciones o conclusiones** sobre una **población** basándonos en los datos de una **muestra** (Esto es fundamental en el análisis de datos, ya que a menudo no es práctico o posible recopilar datos de toda una población.) . Es una de las ramas principales de la estadística y nos permite tomar decisiones informadas sin necesidad de estudiar toda la población. Esto se hace a través de **métodos probabilísticos** que nos ayudan a estimar parámetros desconocidos y a evaluar la fiabilidad de esas estimaciones.

 **Población vs. Muestra**

- **Población**: Es el conjunto completo de todos los elementos o individuos que tienen las características que nos interesa estudiar. 
  - Ejemplo: Todos los estudiantes de una universidad.
- **Muestra**: Es un subconjunto de la población que seleccionamos para hacer un análisis. Debido a que estudiar toda la población puede ser costoso o poco práctico, trabajamos con muestras representativas.
  - Ejemplo: 200 estudiantes seleccionados aleatoriamente de esa universidad.

## **Objetivo de la Inferencia Estadística**

El objetivo principal de la inferencia estadística es hacer **generalizaciones** o **predicciones** sobre la población a partir de una muestra. Estas generalizaciones pueden ser:
- **Estimaciones de parámetros**: Calcular valores que describan la población (como la media, la varianza, la proporción).
- **Pruebas de hipótesis**: Evaluar afirmaciones o suposiciones sobre los parámetros de la población.

##  **Métodos de Inferencia Estadística**

La inferencia estadística se divide en dos áreas principales: **estimación** y **pruebas de hipótesis**.

###  **Estimación de Parámetros**

La **estimación** es el proceso de calcular un valor aproximado de un parámetro desconocido de la población (como la media o la proporción) usando los datos de una muestra.

- **Estimación puntual**: Es un valor único que se usa para estimar un parámetro. Por ejemplo, la media muestral se usa para estimar la media poblacional.
  
  Ejemplo: Si tenemos una muestra de 100 personas y calculamos que la media de edad es 30 años, esa es una estimación puntual de la media de edad de la población.

- **Estimación por intervalo**: Es un rango de valores dentro del cual creemos que se encuentra el parámetro poblacional. A menudo se utiliza el **intervalo de confianza** para hacer este tipo de estimaciones.
  
  Ejemplo: Un intervalo de confianza del 95% para la media de edad de la población podría ser [28, 32] años. Esto significa que estamos 95% seguros de que la media real de la población está entre 28 y 32 años.

###  **Pruebas de Hipótesis**

Las **pruebas de hipótesis** son una parte fundamental de la inferencia estadística y se utilizan para **evaluar** si hay suficiente evidencia en los datos muestrales para **rechazar o no rechazar** una afirmación (hipótesis) sobre un parámetro poblacional.

- **Hipótesis nula ($H_0$)**: Es la afirmación que queremos probar, que generalmente representa la **ausencia de efecto** o una situación de **no diferencia**.
  - Ejemplo: "No hay diferencia en la media de edad entre dos grupos".
  
- **Hipótesis alternativa ($H_1$ o $H_a$)**: Es la afirmación opuesta a la hipótesis nula. Se acepta si hay suficiente evidencia para rechazar la hipótesis nula.
  - Ejemplo: "Hay una diferencia significativa en la media de edad entre dos grupos".
  
- **Nivel de significancia ($\alpha$)**: Es el umbral que determinamos antes de realizar la prueba para decidir si rechazamos la hipótesis nula. Comúnmente se usa un valor de 0.05 (5%), lo que significa que estamos dispuestos a aceptar un 5% de probabilidad de cometer un error tipo 1 (rechazar una hipótesis nula que es verdadera).
  
- **Valor p**: El valor p es una medida que nos indica la probabilidad de observar los datos, o algo más extremo, dado que la hipótesis nula (H0) es verdadera. En otras palabras, el valor p nos ayuda a evaluar la evidencia en contra de la hipótesis nula. Si el valor p es menor que el nivel de significancia ($\alpha$), **rechazamos la hipótesis nula**.
  
##  **Errores en Inferencia Estadística**

En el proceso de inferencia estadística, pueden ocurrir dos tipos de errores:

- **Error tipo I** (Falso positivo): Rechazar la hipótesis nula cuando en realidad es verdadera. Este error ocurre cuando decimos que existe un efecto o diferencia cuando en realidad no lo hay.
  - Ejemplo: Concluir que un medicamento es eficaz cuando en realidad no lo es.

- **Error tipo II** (Falso negativo): No rechazar la hipótesis nula cuando en realidad es falsa. Este error ocurre cuando no detectamos un efecto que realmente existe.
  - Ejemplo: Concluir que un medicamento no tiene efecto cuando en realidad sí lo tiene.

## **Inferencia Estadística en Machine Learning**

En **Machine Learning**, la inferencia estadística es fundamental para evaluar la calidad y fiabilidad de los modelos. A menudo, se usan pruebas de hipótesis y estimaciones para:
- **Evaluar la precisión** de los modelos.
- **Determinar la significancia** de las características (variables) en los modelos.
- **Estimar la incertidumbre** de las predicciones de los modelos.

Por ejemplo, en un modelo de **regresión lineal**:
- Se utilizan pruebas de hipótesis para determinar si los coeficientes del modelo (que representan la relación entre las variables independientes y la variable dependiente) son significativamente diferentes de cero. Esto nos dice si una variable realmente tiene un impacto sobre la predicción o si el efecto es debido al azar.
  
## **Métodos Comunes en Inferencia Estadística**

- **Prueba t de Student**: Se utiliza para comparar medias entre dos grupos (por ejemplo, comparar el rendimiento de dos tratamientos).  
- **ANOVA**: Sirve para comparar las medias entre tres o más grupos.  
- **Regresión**: Es utilizada para modelar la relación entre una variable dependiente y una o más variables independientes, permitiendo hacer inferencias sobre el impacto de las variables independientes.

## **Resumen de la Inferencia Estadística**

- **Estimación**: Se calcula un parámetro poblacional (como la media o la proporción) a partir de una muestra.
- **Pruebas de hipótesis**: Se evalúan afirmaciones sobre un parámetro poblacional, tomando decisiones basadas en la evidencia.
- **Errores estadísticos**: Pueden ocurrir errores tipo I (falsos positivos) o tipo II (falsos negativos).
- **Machine Learning**: La inferencia estadística se usa para evaluar la calidad de los modelos y la relevancia de las características.
  
La inferencia estadística nos ayuda a hacer **conclusiones informadas** basadas en **muestras limitadas** de datos, y es un componente esencial en el análisis de datos, experimentación y modelado.

## **Pasos en la prueba de hipótesis**

## Ejemplo 1

En este primer ejemplo, mostraremos cómo demostrar (o refutar), con evidencia estadística, que el IMC de las mujeres es diferente al de los hombres.

### 1. Elija una estadística de muestra


El primer paso para la prueba de hipótesis es elegir un estadístico de prueba muestral. La prueba de hipótesis permite comparar el estadístico muestral con el de otra muestra o población. Sea 𝜇1 la media poblacional del IMC de los hombres y 𝜇2 la media poblacional del IMC de las mujeres. Compararemos estadísticamente estos valores medios, :$\mu\_{1}$ y $\mu\_{2}$.


### 2. Definir hipótesis (Nula y Alternativa)


El siguiente paso es definir la hipótesis que se va a probar. La hipótesis se define de dos maneras: hipótesis nula e hipótesis alternativa.

**La hipótesis nula** es una hipótesis estadística que supone que la diferencia en las observaciones se debe a un factor aleatorio. Se denota por $H_0$. 

**La hipótesis alternativa** es lo opuesto a la hipótesis nula. Supone que la diferencia en las observaciones es el resultado de un efecto real. La hipótesis alternativa se denota por H1.

$ 𝐻_{0}:\mu_{1}-\mu_{2} = 0 $ No hay diferencia entre el IMC de los hombres y el IMC de las mujeres. \
$ 𝐻_{A}:\mu_{1}-\mu_{2} != 0 $ Hay diferencia entre el IMC de los hombres y el IMC de las mujeres.

El signo igual en la hipótesis nula indica que es una prueba de dos colas.

### 3. Establecer los criterios de decisión


![](https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-ML0232EN-SkillsNetwork/asset/Hypothesis_Testing_3.png)


Para establecer los criterios de decisión en una prueba de hipótesis, es necesario indicar el nivel de significancia, que representa el umbral de tolerancia para cometer un error tipo I (rechazar una hipótesis nula que es verdadera). Comúnmente, se seleccionan niveles de significancia del 5%, 1% o 0.5%. Dependiendo de este nivel, se decide si aceptamos la hipótesis nula o rechazamos la alternativa, o viceversa.

El diagrama anterior ilustra los principios básicos de la prueba de hipótesis. En este caso, elegiremos un nivel de significación del 5%, lo que implica que el valor alfa será 0.05. Dado que estamos realizando una prueba de dos colas, debemos dividir este valor entre 2, lo que nos da un alfa de 0.025 para cada cola. Por lo tanto, si el valor **p** calculado es menor que el valor de alfa, rechazaremos la hipótesis nula en favor de la hipótesis alternativa. El nivel de relevancia o significancia se determina en función de los requisitos específicos del estudio o negocio en cuestión.

Si deseas profundizar en el concepto de significación estadística, puedes consultar este enlace de [Wikipedia](https://es.wikipedia.org/wiki/Significaci%C3%B3n_estad%C3%ADstica).

En este laboratorio, utilizaremos algunas de las principales estadísticas de prueba, como la prueba t, la puntuación z, la puntuación f o la prueba chi-cuadrado, para evaluar nuestros resultados.

- **Prueba t**: Se utiliza para evaluar la media de una población frente a un valor estándar o para comparar las medias de dos poblaciones, especialmente cuando no se conoce la desviación estándar de la población y se dispone de una muestra pequeña ($n < 30$).
  
- **Puntuación z**: Se utiliza cuando se conoce la desviación estándar de la población o cuando se comparan las medias de dos poblaciones con muestras grandes ($n \geq 30$), sin importar si conocemos o no la desviación estándar de las poblaciones. Además, se usa para comparar proporciones de características frente a proporciones estándar o para comparar las proporciones entre dos poblaciones.

- **Prueba F**: Se emplea para comparar las varianzas entre dos poblaciones, independientemente del tamaño de la muestra. Es la base de técnicas como el Análisis de la Varianza (ANOVA).

- **Prueba chi-cuadrado**: Se usa para determinar si existe una diferencia estadísticamente significativa entre las frecuencias observadas y las frecuencias esperadas en una o más categorías de una tabla de contingencia. Esta tabla de contingencia es una representación tabular de datos categóricos y permite ver cómo se distribuyen las frecuencias de las variables.

Si deseas obtener más detalles sobre las estadísticas y cómo se aplican en las pruebas de hipótesis (como la prueba t, puntuación z, puntuación f o chi-cuadrado), puedes consultar los enlaces correspondientes en Wikipedia para cada uno de estos tests.


### 4. Evaluar e interpretar el resultado

Primero, obtengamos todas las observaciones de hembras y machos utilizando la función `loc()`.

```python
data.shape
```

```python
female=data.loc[data.sex=="female"]
male=data.loc[data.sex=="male"]
```

```python
female.shape
```

```python
male.shape
```

Ahora, seleccionemos los valores del IMC para mujeres y hombres.

```python
f_bmi = female.bmi
m_bmi = male.bmi
```

Ahora, graficaremos la distribución de los valores 'bmi' para mujeres y hombres usando la función 'distplot()' de seaborn.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", context="talk")

plt.figure(figsize=(10,6))

sns.kdeplot(f_bmi, color="green", linewidth=2, label="BMI Mujeres", fill=True, alpha=0.3)
sns.kdeplot(m_bmi, color="red", linewidth=2, label="BMI Hombres", fill=True, alpha=0.3)

plt.title("Distribución del BMI por género")
plt.xlabel("BMI")
plt.ylabel("Densidad")
plt.legend()

plt.tight_layout()
plt.show()
```

Desde el gráfico ya vemos que las dos distribuciones son muy similares.

Ahora, calculemos los valores medios del IMC de mujeres y hombres.

```python
female.bmi.mean()
```

```python
male.bmi.mean()
```

A continuación, obtendremos nuestras estadísticas, el valor t y el valor p. Utilizaremos la biblioteca `scipy.stats` y la función `ttest_ind()` para calcular estos parámetros.

```python
alpha = 0.05  
# Nivel de significancia: probabilidad máxima de rechazar la hipótesis nula cuando es verdadera (5%)

t_value1, p_value1 = stats.ttest_ind(m_bmi, f_bmi)
# Realiza una prueba t de Student para dos muestras independientes:
# - m_bmi: BMI de hombres
# - f_bmi: BMI de mujeres
# La prueba compara si las medias de ambos grupos son estadísticamente diferentes
# t_value1: estadístico t calculado
# p_value1: valor p asociado a la prueba

print("t_value1 = ", t_value1, ", p_value1 = ", p_value1)
# Imprime el estadístico t y el valor p
```

A continuación, aunque es opcional, es útil imprimir declaraciones 'if/else' para sacar conclusiones sobre la hipótesis.

```python
if p_value1 <alpha:
    print("Conclusión: dado que p_value {} es menor que alfa {} ". format (p_value1,alpha))
    print("Rechace la hipótesis nula de que no hay diferencia entre el IMC de las mujeres y el IMC de los hombres.")
    
else:
    print("Conclusión: dado que p_value {} es mayor que alfa {} ". format (p_value1,alpha))
    print("No se puede rechazar la hipótesis nula de que existe una diferencia entre el IMC de las mujeres y el IMC de los hombres.")
```

Conclusión: No rechazamos la hipótesis nula y podemos concluir que no hay diferencia entre el IMC de mujeres y hombres.

En este ejemplo, queremos demostrar (o refutar) que las afirmaciones médicas de quienes fuman son mayores que las de quienes no fuman.

Compararemos los valores medios (𝜇) de la población fumadora y no fumadora. Primero, necesitamos calcular los valores medios de la población fumadora y no fumadora.


```python
smoker = data.loc[data.smoker == "yes"]
# Filtra el DataFrame 'data' y selecciona solo las filas donde la columna 'smoker' es igual a "yes" (personas fumadoras)

smoker_char = smoker.charges
# Extrae la columna 'charges' del DataFrame filtrado Contiene los costos médicos de las personas fumadoras

sch_mean = smoker_char.mean()
# Calcula el valor promedio (media) de los costos médicos de las personas fumadoras

sch_mean
# Muestra el valor de la media calculada
```

```python
nonsmoker = data.loc[data.smoker=="no"]
nonsmoker_char = nonsmoker.charges
```

## Ejercicio 1

Calcule la media poblacional de no fumadores.


```python
nsch_mean = nonsmoker_char.mean()
nsch_mean
```

Ahora, definamos nuestra hipótesis nula y alternativa.

$ 𝐻_{0}:\mu_{1}<=\mu_{2} $ Los cargos promedio de los fumadores son menores o iguales a los de los no fumadores. \
$ 𝐻_{A}:\mu_{1}>\mu_{2} $ Los cargos promedio de los fumadores son mayores o iguales a los de los no fumadores.

El signo ">" en la hipótesis alternativa indica que la prueba es de cola derecha. Para comparar las medias de las poblaciones de fumadores y no fumadores, utilizaremos una prueba t. Si los valores z (calculados mediante una prueba t) se encuentran en el área del lado derecho de una curva de distribución, esto nos haría rechazar la hipótesis nula.

Ahora, graficaremos las poblaciones de fumadores y no fumadores usando la función `boxplot()` de *seaborn*. Siempre es útil tener una representación visual de los datos con los que trabajamos.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

sns.set(style="whitegrid", context="talk")

plt.figure(figsize=(8,6))

sns.boxplot(
    x="smoker",
    y="charges",
    data=data,
    palette=["#d95f02", "#1b9e77"]
)

plt.title("Fig. 1: Costos médicos según hábito de fumar")
plt.xlabel("Fumador")
plt.ylabel("Costos médicos")

plt.tight_layout()
plt.show()
```

Ahora, calcularemos el valor t y el valor p de los cargos para las poblaciones de fumadores y no fumadores.

```python
alpha = 0.05  
# Establece el nivel de significancia para la prueba.  Si el valor p es menor que 0.05, rechazaremos la hipótesis nula.

# Realiza una prueba t de Student para dos muestras independientes: 'smoker_char' y 'nonsmoker_char'
# La prueba compara las medias de los costos médicos entre fumadores y no fumadores
t_val2, p_value2 = stats.ttest_ind(smoker_char, nonsmoker_char)  
# t_val2: estadístico t calculado (mide la diferencia entre las medias, en términos de desviaciones estándar)
# p_value2: valor p calculado (probabilidad de que la diferencia observada sea debido al azar)

# Calcula el valor p para una cola (unilateral), dividiendo el p_value entre 2
p_value_onetail = p_value2 / 2  
# Esto es útil si la hipótesis alternativa especifica una dirección (por ejemplo, "los fumadores tienen costos mayores").

# Imprime los resultados de la prueba t
print("t_value = {} , p_value ={} , p_value_onetail = {}".format(t_val2, p_value2, p_value_onetail))  
# Muestra el valor t, el valor p para dos colas y el valor p para una cola (unilateral)
```

## Ejercicio 2

Utilice la función `print()` para establecer sus conclusiones basándose en las estadísticas calculadas. ¿Cuáles son las conclusiones?


```python
if p_value2 < alpha:
    # Si el valor p es menor que el nivel de significancia (alpha), rechazamos la hipótesis nula
    print(f"Conclusión: Dado que p_value {p_value_onetail:.6f} es menor que alpha {alpha},")
    print("Rechazamos la hipótesis nula de que los cargos promedio para los fumadores son menores o iguales a los de los no fumadores.")
else:
    # Si el valor p es mayor que alpha, no rechazamos la hipótesis nula
    print(f"Conclusión: Dado que p_value {p_value_onetail:.6f} es mayor que alpha {alpha},")
    print("No rechazamos la hipótesis nula de que los cargos promedio para los fumadores son menores o iguales a los de los no fumadores.")
```

# Niveles de significancia y valores p

El nivel de significancia establece un umbral para decidir si los resultados son estadísticamente significativos

Para obtener un **motivo de rechazo**, debemos calcular la **estadística de prueba**.

El **nivel de significancia** ($\alpha$) está estrechamente relacionado con la importancia de evitar errores de tipo 1 y tipo 2. Un **error de tipo 1** ocurre cuando rechazamos incorrectamente una hipótesis nula verdadera, mientras que un **error de tipo 2** ocurre cuando no rechazamos una hipótesis nula falsa. 

La idea detrás de un **nivel de significancia más bajo** es que solo rechazaríamos la hipótesis nula si la probabilidad de los datos observados (o más extremos) es extremadamente baja, lo cual aumenta nuestra confianza de que los resultados no ocurrieron por azar. 

Por lo tanto, es esencial **elegir un valor de $\alpha$ antes de realizar el análisis estadístico**. Si no lo hacemos, corremos el riesgo de cometer *p-hacking*, un comportamiento en el que manipulamos los datos (como modificar el valor p) para que la hipótesis nula se rechace. Esto puede llevar a conclusiones erróneas y a una mala interpretación de los resultados.

Es crucial que al definir el nivel de significancia, tengamos en cuenta el **balance entre evitar un error de tipo 1** (rechazar incorrectamente la hipótesis nula) y **el riesgo de un error de tipo 2** (no rechazar una hipótesis nula falsa). Elegir un valor adecuado de $\alpha$ es fundamental para realizar un análisis robusto y fiable.

|------------------------|------------------------------|--------------------------|
| **Rechazar la H0**      | Error de Tipo 1 (Falso positivo) | Correcto (Verdadero positivo) |
| **No rechazar la H0**   | Correcto (Verdadero negativo)   | Error de Tipo 2 (Falso negativo) |


## P-valor y su Interpretación

* El **p-valor** es la probabilidad bajo la distribución nula de obtener un resultado tan extremo como el que realmente se observó, o más extremo, si la hipótesis nula fuera cierta. En otras palabras, el p-valor nos ayuda a determinar si nuestros datos son consistentes con la hipótesis nula o si es razonable rechazarla.

  - Si el p-valor es **pequeño** (típicamente menor a un nivel de significancia, como 0.05), indica que los resultados observados son **muy improbables** bajo la hipótesis nula, por lo que generalmente **rechazamos la hipótesis nula**.
  - Si el p-valor es **grande**, significa que los datos observados son consistentes con la hipótesis nula y, por lo tanto, **no rechazamos la hipótesis nula**.

  **Ejemplo de la vida diaria:**
  
  Imagina que estás probando si un nuevo medicamento tiene efectos positivos en la salud. La hipótesis nula sería que el medicamento no tiene efecto, y la hipótesis alternativa sería que sí lo tiene. Si al analizar los datos del experimento, obtienes un p-valor de 0.03, esto indica que hay un 3% de probabilidad de observar estos resultados (o algo más extremo) si el medicamento realmente no tiene efecto. Debido a que este p-valor es bajo, puedes rechazar la hipótesis nula y concluir que el medicamento tiene un efecto positivo.



## Intervalo de Confianza y su Relación con la Hipótesis Nula

* El **intervalo de confianza** es un rango de valores en el cual **esperamos** que se encuentre el valor verdadero de una estimación con un cierto nivel de confianza (por ejemplo, 95%). Este intervalo se calcula a partir de los datos muestrales y ayuda a indicar la precisión de la estimación.

  - Si el intervalo de confianza **no incluye** el valor de la hipótesis nula (por ejemplo, 0 en una prueba de diferencia de medias), esto sugiere que los resultados son **significativos** y, por lo tanto, la hipótesis nula podría ser rechazada.
  - Si el intervalo de confianza **incluye** el valor de la hipótesis nula, no podemos rechazarla porque los resultados observados podrían ser el resultado de un efecto nulo (sin efecto real).

  **Ejemplo de la vida diaria:**

Después de realizar el análisis, calculas un intervalo de confianza del 95% para la media, y obtienes el rango de [6.2, 7.8]. Como este intervalo **incluye 7**, no puedes rechazar la hipótesis nula (que es que la media es 7), ya que los resultados no son lo suficientemente consistentes para asegurar que la media es diferente.



### Resumen: P valor e Intervalo de confianza

- **P-valor**: El **p-valor** es una medida estadística que nos indica **qué tan compatible** es la evidencia observada (en los datos) con una hipótesis nula, bajo la suposición de que dicha hipótesis es verdadera. ## La F-estadística

La **F-estadística** es una medida que se utiliza en las pruebas estadísticas para comparar la variabilidad entre grupos con la variabilidad dentro de los grupos. Se usa principalmente en el análisis de la **varianza** (ANOVA) y otras pruebas de comparación de modelos. La F-estadística es el cociente de dos estimaciones de la varianza, lo que nos da una idea de si los grupos tienen variaciones similares o si hay diferencias significativas entre ellos.

#### Fórmula de la F-estadística

La **F-estadística** se calcula de la siguiente forma:

$$
F = \frac{\text{Varianza entre los grupos}}{\text{Varianza dentro de los grupos}}$$

- **Varianza entre los grupos**: Mide la variabilidad de las medias de los diferentes grupos con respecto a la media global.
- **Varianza dentro de los grupos**: Mide la variabilidad de las observaciones dentro de cada grupo.

#### ¿Qué significa el valor de la F-estadística?

- **F alto**: Si el valor de la F-estadística es **alto**, sugiere que la variabilidad entre los grupos es mucho mayor que la variabilidad dentro de los grupos. Esto generalmente indica que al menos uno de los grupos es significativamente diferente de los otros. En otras palabras, existe **una diferencia significativa** entre las medias de los grupos.
  
- **F bajo**: Si el valor de la F-estadística es **bajo**, sugiere que la variabilidad entre los grupos es similar a la variabilidad dentro de los grupos. Esto significa que **no hay diferencias significativas** entre las medias de los grupos, y la hipótesis nula de igualdad de medias se mantiene.

#### Ejemplo de la vida diaria:

Imagina que estás probando tres métodos diferentes para enseñar matemáticas a estudiantes, y quieres saber si hay una diferencia significativa en las calificaciones de los estudiantes entre los métodos. Realizas un análisis de varianza (ANOVA) para comparar los tres grupos de estudiantes.

- Si la **F-estadística** resultante es **alta**, eso sugiere que al menos uno de los métodos de enseñanza tiene un impacto significativamente diferente en las calificaciones.
- Si la **F-estadística** es **baja**, entonces podrías concluir que los métodos de enseñanza no tienen diferencias significativas en cuanto a sus efectos en las calificaciones.

#### Interpretación del valor p asociado con la F-estadística

- Si el valor de la **F-estadística** es suficientemente alto, el valor **p** asociado será **bajo**, lo que sugiere que puedes rechazar la hipótesis nula y concluir que hay una diferencia significativa entre los grupos.
- Si el valor de la **F-estadística** es bajo, el valor **p** será alto, lo que sugiere que no hay suficiente evidencia para rechazar la hipótesis nula.

#### Resumen

La **F-estadística** es fundamental en la prueba de hipótesis cuando se comparan varias medias o grupos. Nos permite evaluar si las diferencias observadas en los datos son lo suficientemente grandes como para ser consideradas significativas o si podrían haber ocurrido por casualidad. Este valor es especialmente útil en el **Análisis de Varianza (ANOVA)**, que es común en estudios experimentales donde se comparan más de dos grupos.




## F-estadística 

En el contexto de **Machine Learning**, la **F-estadística** se utiliza principalmente para evaluar la **importancia** de las características (o variables) en modelos estadísticos, como en el análisis de regresión o en el análisis de la varianza (ANOVA). Se utiliza para determinar si la **variabilidad** de los resultados de un modelo puede explicarse significativamente por las características de entrada, o si es simplemente ruido o azar.

### ¿Cómo se aplica la F-estadística en Machine Learning?

En Machine Learning, la F-estadística suele aparecer cuando se usan **modelos de regresión** (especialmente en regresión lineal) y en **ANOVA** para determinar la significancia de las variables.

1. **En Regresión Lineal**: 
   - En la regresión lineal, la F-estadística evalúa si el modelo, en general, es significativo. Es decir, si las características de entrada como grupo tienen una relación significativa con la variable de salida. La hipótesis nula en este caso es que **ninguna de las variables predice significativamente** la variable dependiente.
   - Si el valor de la F-estadística es alto, indica que al menos una de las variables tiene una relación significativa con la variable objetivo, lo que sugiere que el modelo es relevante.
   
   **Fórmula de la F-estadística en regresión lineal**:
   $$
   F = \frac{(\text{Suma de cuadrados explicada} / \text{Grados de libertad explicados})}{(\text{Suma de cuadrados residual} / \text{Grados de libertad residuales})}
   $$
   - **Suma de cuadrados explicada (SSR)**: Mide la variabilidad de los datos que es explicada por el modelo.
   - **Suma de cuadrados residual (SSE)**: Mide la variabilidad de los datos que no es explicada por el modelo.

2. **En ANOVA** (Análisis de Varianza):
   - **ANOVA** se utiliza para comparar las medias de tres o más grupos para ver si existen diferencias significativas entre ellas. En este contexto, la F-estadística compara la variabilidad entre los grupos con la variabilidad dentro de los grupos.
   - Si el valor de la F-estadística es alto, sugiere que al menos uno de los grupos tiene una media significativamente diferente a los demás, lo que podría indicar que una variable predictora tiene un impacto importante en la salida del modelo.
   
   **Ejemplo de uso de F-estadística en Machine Learning**:
   Imagina que estás utilizando un **modelo de regresión lineal múltiple** para predecir los precios de las casas con características como el tamaño, la ubicación, el número de habitaciones, etc. La **F-estadística** se usaría para evaluar si las variables que has seleccionado (tamaño, ubicación, número de habitaciones) en conjunto son útiles para predecir el precio de la casa.
   
   - Si el valor de la F-estadística es **alto**, indica que las características seleccionadas explican bien la variabilidad en los precios, lo que sugiere que el modelo es útil.
   - Si el valor de la F-estadística es **bajo**, sugiere que las características seleccionadas no aportan mucho al modelo y que puede que debas buscar otras variables o ajustar tu modelo.

### Relación con el valor p

- Al igual que en otros contextos estadísticos, el valor **p** asociado con la F-estadística te indica si puedes **rechazar la hipótesis nula** de que no hay relación entre las variables. 
- En el contexto de **Machine Learning**, si el valor **p** asociado con la F-estadística es **bajo** (típicamente menor a 0.05), puedes concluir que al menos una de las características tiene un impacto significativo en el modelo.
  
### Ejemplo en un Modelo de Regresión Lineal:

Supón que has entrenado un **modelo de regresión lineal múltiple** para predecir el precio de una casa en función de las características como el número de habitaciones, la ubicación y el tamaño en metros cuadrados.

- Al realizar un análisis, obtienes una F-estadística de 25.5 con un valor p de 0.003. Esto significa que la probabilidad de obtener un valor tan extremo como 25.5 bajo la hipótesis nula (de que las características no tienen efecto) es muy baja. Así que, **rechazas la hipótesis nula** y concluyes que al menos una de las características tiene un impacto significativo en el precio de la casa.

### ¿Por qué es importante en Machine Learning?

La **F-estadística** y el valor p asociado te ayudan a:
- **Evaluar la calidad del modelo**: Determinar si el modelo tiene un buen rendimiento y si las variables seleccionadas son realmente relevantes.
- **Seleccionar características**: Ayuda a decidir qué características (o variables) incluir en el modelo. Si una característica tiene una F-estadística baja, podrías considerar eliminarla del modelo.
- **Mejorar la interpretabilidad**: A través de la F-estadística, puedes entender qué variables están teniendo el mayor impacto en tu modelo.

### Resumen

En **Machine Learning**, la **F-estadística** es utilizada para evaluar la significancia de las variables de entrada en modelos estadísticos como la regresión lineal o el análisis de varianza (ANOVA). Si la F-estadística es alta, sugiere que las variables seleccionadas explican bien la variabilidad en la variable dependiente, lo que indica que el modelo es efectivo. También se utiliza para la **selección de características** y la **evaluación de la calidad del modelo** en general.



## Power y Tamaño de la Muestra (Sample Size) 

En estadísticas y **Machine Learning**, el **power** (potencia) y el **tamaño de la muestra** son dos conceptos fundamentales cuando se diseñan experimentos y pruebas estadísticas. Ambos conceptos están relacionados con la capacidad de una prueba para detectar efectos reales y la cantidad de datos necesarios para obtener resultados significativos y confiables.

###  **Power (Potencia)**

El **power** de una prueba estadística es la probabilidad de **rechazar correctamente la hipótesis nula** cuando esta es falsa. Es decir, el poder mide la **capacidad de una prueba para detectar un efecto real** (si es que existe). En otras palabras, es la probabilidad de evitar un **error de tipo 2** (falso negativo), que es cuando no se rechaza la hipótesis nula aunque la hipótesis alternativa sea verdadera.

- **Poder alto**: Un poder alto significa que la prueba es capaz de detectar efectos significativos en los datos si existen.
- **Poder bajo**: Un poder bajo implica que es menos probable detectar efectos verdaderos, lo que podría llevar a cometer **errores de tipo 2** (no detectar un efecto cuando realmente existe).

La potencia de una prueba depende de varios factores:
- **Tamaño de la muestra**: Cuanto mayor sea el tamaño de la muestra, mayor será la potencia.
- **Magnitud del efecto**: Si el efecto que estás tratando de detectar es grande, el poder será más alto.
- **Nivel de significancia** ($\alpha$): Si se utiliza un nivel de significancia más bajo (por ejemplo, 0.01), el poder será más bajo, ya que es más difícil rechazar la hipótesis nula.
- **Variabilidad en los datos**: Menos variabilidad (más precisión en los datos) aumenta el poder.

###  **Tamaño de la Muestra (Sample Size)**

El **tamaño de la muestra** es el número de observaciones o datos que se recogen para un experimento. El tamaño de la muestra tiene un impacto directo sobre la potencia de una prueba estadística:

- **Muestras pequeñas**: Con una muestra pequeña, es más difícil detectar una diferencia significativa, incluso si esta existe (baja potencia).
- **Muestras grandes**: Con un tamaño de muestra más grande, se tiene una mayor capacidad para detectar diferencias significativas, lo que aumenta la potencia de la prueba.

#### Relación entre Power y Tamaño de la Muestra

El tamaño de la muestra y el poder están estrechamente relacionados. Un mayor tamaño de muestra generalmente aumenta la potencia de la prueba, lo que mejora las posibilidades de **detectar un efecto real** cuando este existe.

**Ejemplo práctico:**
Imagina que estás entrenando un modelo de Machine Learning para predecir si un paciente tiene una enfermedad en función de sus características. Si tienes solo 50 muestras (pacientes), es posible que el poder de tu prueba sea bajo, lo que significa que podrías no detectar una diferencia significativa entre los grupos (enfermos vs. sanos). Si aumentas el tamaño de la muestra a 500 pacientes, tendrás una mayor capacidad para detectar patrones en los datos y el poder de tu prueba aumentará.

###  **Cálculo del Power y Tamaño de la Muestra**

El cálculo del poder y el tamaño de la muestra es crucial antes de realizar un experimento o análisis de datos. En Machine Learning, esto se hace para asegurarse de que los resultados sean significativos y no producto del azar. Existen herramientas y métodos como el **Análisis de Potencia** que te ayudan a determinar:

- Cuántos datos necesitas para detectar un efecto específico con un cierto nivel de confianza.
- Qué tan grande debe ser tu muestra para tener un poder suficiente (generalmente, se busca un poder de 0.80, es decir, un 80% de probabilidad de detectar un efecto verdadero si existe).

Los cálculos de tamaño de muestra suelen requerir información sobre:
- El **efecto esperado**: Cuán grande crees que será la diferencia entre los grupos.
- El **nivel de significancia** ($\alpha$): Usualmente 0.05 o 0.01.
- El **poder deseado**: Por lo general, se busca un poder de al menos 0.80.

Existen herramientas estadísticas y software (como **G*Power**, **R** o **Python**) que pueden ayudarte a calcular estos valores antes de realizar un análisis.

Si solo tienes 100 muestras, puede ser difícil detectar una relación significativa entre las características y la compra, debido al tamaño pequeño de la muestra. Esto podría resultar en un **error de tipo 2**, donde el modelo no detecta correctamente los patrones en los datos.

Si decides aumentar el tamaño de la muestra a 1,000 o 10,000 clientes, el poder de la prueba aumentará. Esto te permitirá **detectar relaciones más pequeñas** entre las características y la compra del producto, mejorando la capacidad de tu modelo para hacer predicciones más precisas.

## Corrección de Bonferroni

La **corrección de Bonferroni** es una técnica estadística utilizada para **ajustar el nivel de significancia** cuando se realizan múltiples pruebas estadísticas simultáneamente. Esta corrección ayuda a reducir el **riesgo de cometer errores de tipo 1** (falsos positivos) cuando realizamos muchas comparaciones.

**¿Por qué es necesaria la corrección de Bonferroni?**

Cuando realizamos **múltiples pruebas estadísticas**, cada una de ellas tiene una cierta probabilidad de generar un **error de tipo 1** (rechazar incorrectamente una hipótesis nula verdadera). Si realizamos muchas pruebas sin ajustarnos, el riesgo de obtener al menos un falso positivo aumenta, incluso si las hipótesis nulas son verdaderas.

La **corrección de Bonferroni** ajusta el nivel de significancia para compensar el aumento en el riesgo de cometer errores de tipo 1 cuando se hacen múltiples pruebas.

**Fórmula de la Corrección de Bonferroni**

La corrección de Bonferroni se realiza ajustando el nivel de significancia ($\alpha$) en función del número de pruebas que se realicen:

$$
\alpha_{\text{ajustado}} = \frac{\alpha}{m}
$$

Donde:
- **$\alpha$** es el nivel de significancia original (por ejemplo, 0.05).
- **$m$** es el número de pruebas estadísticas realizadas.

**¿Cómo funciona la Corrección de Bonferroni?**

La idea detrás de la corrección de Bonferroni es que, al realizar múltiples pruebas, el nivel de significancia global debe ser **más estricto** para que el riesgo de cometer un error de tipo 1 no se acumule. Al dividir el nivel de significancia original ($\alpha$) por el número de pruebas, se reduce el umbral necesario para rechazar una hipótesis nula.

**Ejemplo práctico:**

Supón que realizas **5 pruebas estadísticas** en un conjunto de datos y usas un nivel de significancia de **0.05**. Sin corrección, el riesgo de cometer un error de tipo 1 en al menos una de las pruebas es 0.05 por prueba, lo que podría ser mayor que 0.05 debido a la acumulación de pruebas.

Para aplicar la **corrección de Bonferroni**, ajustamos el valor de $\alpha$:

$$
\alpha_{\text{ajustado}} = \frac{0.05}{5} = 0.01
$$

Esto significa que, después de la corrección, solo podrás rechazar una hipótesis nula si el valor p es menor a **0.01** en lugar de **0.05**.

**¿Cuándo utilizar la corrección de Bonferroni?**

La corrección de Bonferroni es especialmente útil en los siguientes escenarios:
- Cuando realizas muchas pruebas estadísticas sobre el mismo conjunto de datos y no deseas que el riesgo de obtener un falso positivo sea excesivo.
- En estudios donde es crucial **minimizar los falsos positivos**, como en estudios clínicos o experimentos científicos.

**Desventajas de la Corrección de Bonferroni**

Aunque la corrección de Bonferroni es muy efectiva para controlar el **error tipo 1**, tiene algunas desventajas importantes:

- **Reducción de potencia**: Al hacer que el umbral de significancia sea más estricto, aumentamos la probabilidad de cometer **errores de tipo 2** (falsos negativos). Es decir, puede ser más difícil detectar efectos reales cuando se aplica esta corrección, especialmente cuando se realizan muchas pruebas.
- **Conservadora**: Es una corrección conservadora, lo que significa que puede ser demasiado estricta cuando se realizan pocas pruebas. En estos casos, puede que estés perdiendo la oportunidad de detectar efectos reales.

**Ejemplo en Machine Learning**

Imagina que estás entrenando un modelo de **regresión múltiple** para predecir el rendimiento de estudiantes en función de varias características, como horas de estudio, participación en clase, etc. Realizas pruebas de hipótesis para cada una de las variables (características) para ver si influyen significativamente en la predicción.

Si realizas, por ejemplo, **10 pruebas estadísticas**, el riesgo de obtener al menos un falso positivo aumenta si no corriges el nivel de significancia. Usando la **corrección de Bonferroni**, divides tu nivel de significancia original (por ejemplo, 0.05) entre el número de pruebas (10), lo que ajusta el umbral para cada prueba individual y reduce la probabilidad de cometer un falso positivo.

**Resumen**

- **Corrección de Bonferroni** ajusta el nivel de significancia cuando se realizan múltiples pruebas estadísticas para controlar el riesgo de cometer errores tipo 1 (falsos positivos).
- La fórmula es: $$\alpha_{\text{ajustado}} = \frac{\alpha}{m}$$, donde **$\alpha$** es el nivel de significancia original y **$m$** es el número de pruebas realizadas.
- **Ventajas**: Controla el riesgo de cometer falsos positivos cuando se hacen múltiples comparaciones.
- **Desventajas**: Puede reducir la potencia de la prueba, aumentando la probabilidad de cometer errores de tipo 2 (falsos negativos).

Es útil cuando se están evaluando múltiples hipótesis simultáneamente, pero siempre hay que considerar el **equilibrio entre el control de falsos positivos y la detección de efectos reales**.



### Correlación vs Causalidad

En estadísticas y **Machine Learning**, es fundamental comprender la diferencia entre **correlación** y **causalidad**, ya que son conceptos que a menudo se confunden, pero tienen significados muy diferentes.

### **Correlación**

La **correlación** se refiere a la **relación estadística** entre dos o más variables. Cuando dos variables están correlacionadas, significa que los cambios en una variable están asociados con cambios en otra, pero **no necesariamente** indican que una causa a la otra.

- **Correlación positiva**: Cuando una variable aumenta, la otra también tiende a aumentar. Ejemplo: A medida que aumenta la cantidad de horas estudiadas, las calificaciones de los estudiantes tienden a mejorar.
- **Correlación negativa**: Cuando una variable aumenta, la otra tiende a disminuir. Ejemplo: A medida que el número de horas de sueño disminuye, el nivel de concentración en el trabajo o estudio puede disminuir.
  
**Coeficiente de correlación de Pearson**: Una de las formas más comunes de medir la correlación es mediante el coeficiente de correlación de Pearson, que tiene un valor entre -1 y 1:
- **1** indica una correlación positiva perfecta.
- **-1** indica una correlación negativa perfecta.
- **0** indica **ausencia de correlación**.

###  **Causalidad**

La **causalidad** implica que un cambio en una variable **provoca un cambio directo** en otra variable. La causalidad es una **relación de causa y efecto** en la que la **causa precede al efecto**.

- **Causalidad**: Si se establece que A causa B, significa que el cambio en A directamente provoca el cambio en B. Por ejemplo, **fumar (A) causa (B) cáncer de pulmón**.
  
En estudios experimentales, la causalidad se puede probar si se manipula una variable (la causa) y se observa el efecto en la otra variable (el efecto).

 **¿Por qué Correlación no implica Causalidad?**

El dicho popular **"correlación no implica causalidad"** destaca un concepto importante: **una correlación entre dos variables no significa que una cause la otra**. Pueden ocurrir varias razones por las cuales dos variables están correlacionadas, pero no hay una relación causal directa:

- **Coincidencia o azar**: A veces, dos variables pueden estar correlacionadas por pura casualidad, sin que una cause a la otra.
- **Tercera variable**: Una **variable de confusión** puede estar influyendo en ambas variables, haciendo que parezca que están correlacionadas cuando en realidad una tercera variable es la que está causando ambas.
  
**Ejemplo clásico:**
- Existe una correlación positiva entre el número de helados vendidos y el número de ahogamientos en verano. Esto no significa que comprar helados cause ahogamientos. La causa real es que ambos factores aumentan debido al **calor del verano**, una tercera variable.

 **Ejemplos de Correlación vs Causalidad**

- **Correlación**: Un estudio muestra que las personas que tienen más años de experiencia en un trabajo tienden a tener mayores salarios. Esto puede ser cierto, pero **no significa necesariamente que tener más años de experiencia cause salarios más altos**. Otras variables, como el tipo de trabajo, la industria o la educación, pueden influir en ambos.
  
- **Causalidad**: En los estudios de **medicina**, como los realizados con ensayos clínicos, se puede establecer que un medicamento causa una mejora en la salud de los pacientes, si se observa un efecto significativo en el grupo tratado en comparación con un grupo de control (sin el tratamiento).

 **¿Cómo establecer causalidad?**

Para demostrar causalidad, se necesitan **experimentos controlados** o estudios donde se manipula una variable y se observa el efecto en otra. Algunas formas de establecer causalidad incluyen:

1. **Experimentos controlados aleatorios (ECA)**: Son el estándar en muchos campos, como la medicina. Los participantes se asignan aleatoriamente a diferentes grupos (por ejemplo, un grupo con tratamiento y un grupo de control sin tratamiento) para observar el efecto.
   
2. **Estudios de intervención**: Modificar una variable (por ejemplo, introducir una nueva política educativa) y observar el efecto sobre una variable dependiente (por ejemplo, el rendimiento de los estudiantes).

3. **Modelos estadísticos avanzados**: En algunos casos, se utilizan **modelos estadísticos** como **modelos de variables instrumentales** o **regresión de discontinuidad** para intentar inferir causalidad en situaciones donde no es posible realizar un experimento controlado.

 **Correlación y Causalidad en Machine Learning**

En **Machine Learning**, la **correlación** entre variables es útil para **seleccionar características** relevantes para un modelo, pero **no garantiza causalidad**. Los modelos de Machine Learning, como los de regresión o clasificación, pueden aprender patrones a partir de las correlaciones entre características y la variable objetivo, pero no necesariamente indican que una característica cause el resultado.

**Ejemplo en Machine Learning:**
Imagina que estás construyendo un modelo para predecir el riesgo de enfermedades cardíacas basándote en características como el nivel de colesterol, la edad, el ejercicio, etc. Aunque la **correlación** entre el nivel de colesterol y el riesgo de enfermedades cardíacas puede ser fuerte, esto no necesariamente implica que el colesterol **cause** las enfermedades cardíacas. Otros factores, como la genética o el estilo de vida, pueden ser causas subyacentes.

 **Resumen**

- **Correlación**: Es una relación estadística entre dos variables, donde los cambios en una variable están asociados con los cambios en otra. No implica que una variable cause a la otra.
- **Causalidad**: Implica una relación de causa y efecto, donde un cambio en una variable causa un cambio directo en otra.
- **Correlación no implica causalidad**: La correlación puede ocurrir por azar o por una **variable de confusión**, y no necesariamente significa que una variable cause a la otra.
- **Establecer causalidad**: Se requiere de **experimentos controlados**, **estudios de intervención** o métodos estadísticos avanzados para inferir causalidad.

En **Machine Learning**, es importante no asumir causalidad solo por la correlación y ser consciente de que los modelos pueden aprender patrones sin establecer relaciones causales directas.
