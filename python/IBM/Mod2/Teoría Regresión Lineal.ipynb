## **Regresión lineal y múltiple**

En un modelo de regresión lineal, la relación entre la **variable dependiente** (lo que queremos predecir) y las **variables independientes** (las características o predictoras) se establece mediante una **combinación lineal** de estas últimas.

**Ecuación de regresión lineal simple (una variable independiente)**

Cuando tenemos una sola variable independiente $x$, la relación se describe mediante la siguiente ecuación:

$$
\hat{y} = \beta_0 + \beta_1 \cdot x + \epsilon
$$

Donde:
- $\hat{y} $ es la **variable dependiente** (lo que estamos prediciendo).
- $x$ es la **variable independiente** (la característica que usamos para predecir $y$).
- $\beta_0$ es el **intercepto** o constante. Representa el valor de $y$ cuando $x = 0$.
- $\beta_1$ es el **coeficiente de regresión** o **pendiente**. Indica cuánto cambia $y$ por cada unidad que cambia $x$.
- $\epsilon$ es el **término de error** o residuo, que representa la diferencia entre el valor real observado y el valor predicho por el modelo.

 **Regresión lineal múltiple (varias variables independientes)**

Cuando tenemos más de una variable independiente, la ecuación se generaliza a la **regresión lineal múltiple**:

$$
\hat{y}  = \beta_0 + \beta_1 \cdot x_1 + \beta_2 \cdot x_2 + \dots + \beta_n \cdot x_n + \epsilon
$$

En este caso:
- $x_1, x_2, \dots, x_n$ son las **variables independientes** (por ejemplo, características como tamaño, edad, ingresos, etc.).
- $\beta_1, \beta_2, \dots, \beta_n$ son los **coeficientes de regresión** correspondientes a cada variable independiente.
- $\beta_0$ es el **intercepto**.
- $\epsilon$ es el **error** o residuo.

**Interpretación de los coeficientes**

- **$\beta_0$**: Representa el valor de $y$ cuando todas las $x$'s son igual a cero. A veces no tiene sentido en el contexto de los datos, pero es necesario para el modelo.
- **$\beta_1, \beta_2, \dots, \beta_n$**: Cada coeficiente indica cuán fuerte es la relación entre esa variable independiente y la variable dependiente. Por ejemplo:
  - Si $\beta_1 = 2$, entonces por cada incremento de 1 en $x_1$, el valor de $y$ aumentará en 2, manteniendo las otras variables constantes.
  - Si $\beta_2 = -3$, entonces por cada incremento de 1 en $x_2$, el valor de $y$ disminuirá en 3, manteniendo constantes las demás variables.

 **Ajuste de los coeficientes**

Los valores de $\beta_0, \beta_1, \dots, \beta_n$ se **ajustan** durante el proceso de entrenamiento del modelo. El algoritmo de regresión lineal ajusta estos coeficientes para minimizar el **error cuadrático medio** (ECM), es decir, la diferencia entre los valores reales $y$ y los valores predichos $\hat{y}$.

 **Resumen**

- En regresión lineal, la relación entre la **variable dependiente** $y$ y las **variables independientes** $x_1, x_2, \dots, x_n$ se establece mediante una **combinación lineal** de estas variables.
- El modelo ajusta los **coeficientes** $\beta_1, \beta_2, \dots, \beta_n$ para encontrar la relación más precisa posible entre $y$ y $x_1, x_2, \dots, x_n$, minimizando el error.
- Estos coeficientes nos dicen cómo afecta cada variable independiente a la variable dependiente.
- La regresión lineal busca minimizar la suma de los errores al cuadrado entre las predicciones y los valores reales.

**Ejemplo práctico**

Supongamos que estamos prediciendo el **precio de una casa** ($y$) basado en su **tamaño** ($x_1$) y su **edad** ($x_2$):

$$
\hat{y} = \beta_0 + \beta_1 \cdot x_1 + \beta_2 \cdot x_2 + \epsilon
$$

Si $\beta_1 = 100$, significa que por cada metro cuadrado adicional, el precio de la casa aumenta en 100 unidades de moneda.

Si $\beta_2 = -50$, significa que por cada año adicional de edad de la casa, el precio disminuye en 50 unidades de moneda.

# MSE

El **Error Cuadrático Medio (MSE)** es una métrica utilizada para evaluar el desempeño de un modelo de regresión. Mide el promedio de los **errores cuadrados** entre las predicciones del modelo y los valores reales.

##  **Definición de MSE**

El **MSE** se calcula como la media de los cuadrados de las diferencias entre los valores observados (reales) y los valores predichos por el modelo. La fórmula es la siguiente:

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

Donde:
- $y_i$ es el valor real de la variable dependiente para el i-ésimo dato.
- $\hat{y}_i$ es el valor predicho por el modelo para el i-ésimo dato.
- $n$ es el número total de datos (o puntos de muestra).

 **¿Cómo funciona el MSE?**

El MSE mide cuánto se alejan, en promedio, las predicciones del modelo de los valores reales. La **diferencia entre los valores reales** ($y_i$) y los valores predichos ($\hat{y}_i$) se llama **error**. Si el modelo tiene **predicciones exactas**, el MSE será igual a **0**, ya que no hay error entre los valores reales y las predicciones.

El hecho de **cuadrar los errores** (elevar al cuadrado la diferencia entre los valores reales y los predichos) tiene dos propósitos principales:
1. **Evitar que los errores positivos y negativos se cancelen entre sí**: Al elevar al cuadrado, todos los errores se vuelven positivos, lo que da una medición más consistente de la magnitud del error.
2. **Penalizar más los errores grandes**: Los errores más grandes tienen un mayor impacto en el valor final de MSE, lo que hace que el modelo trate de minimizar estos errores más grandes.

**Importancia del MSE en la evaluación de modelos de regresión**

El MSE es una de las métricas más comunes para evaluar modelos de regresión, ya que tiene las siguientes ventajas:
- **Fácil de calcular**: Es simple y directo de calcular con la fórmula mencionada.
- **Sensibilidad a grandes errores**: Dado que se elevan al cuadrado los errores, el MSE penaliza más los errores grandes. Esto puede ser útil en situaciones donde prefieres evitar grandes desviaciones en las predicciones.
- **Escala proporcional a la magnitud de los datos**: El MSE tiene unidades al cuadrado, lo que significa que la magnitud del MSE está relacionada con las unidades de los datos, lo cual es relevante cuando se comparan modelos con diferentes escalas de datos.

## **Root Mean Squared Error (RMSE)**

El **Error Cuadrático Medio** puede ser difícil de interpretar directamente debido a que está en unidades al cuadrado. Para hacerlo más interpretable, a veces se utiliza la raíz cuadrada del MSE, conocida como el **RMSE (Root Mean Squared Error)**, que devuelve el error a las unidades originales de la variable dependiente:

$$
RMSE = \sqrt{MSE}
$$

 **¿Cuándo usar el MSE?**

El MSE es útil cuando:
- Quieres un modelo que minimice las grandes desviaciones en las predicciones.
- Estás trabajando con datos en los que **los errores grandes son más críticos**.
  
Sin embargo, es importante tener en cuenta que, si el MSE es muy bajo, el modelo está ajustando muy bien los datos, pero si es muy alto, el modelo no está haciendo un buen trabajo al predecir la variable dependiente.

 **Resumen**

El **Error Cuadrático Medio (MSE)** es una medida fundamental para evaluar el rendimiento de un modelo de regresión. Proporciona una forma de medir la diferencia entre los valores reales y los predichos, penalizando los grandes errores. Es útil para:
- Evaluar la precisión de las predicciones.
- Comparar la calidad de diferentes modelos.
- Determinar cuán bien el modelo se ajusta a los datos.

Sin embargo, también es importante considerar otras métricas como el **RMSE** o el **Error Absoluto Medio (MAE)** para obtener una evaluación más completa del modelo.


# Regularización

La **regularización** es un conjunto de técnicas utilizadas para evitar el sobreajuste (overfitting) en los modelos de regresión. El objetivo principal de la regularización es **penalizar** los coeficientes del modelo para que no se ajusten demasiado a los datos de entrenamiento, lo que podría generar un modelo que no generalice bien a nuevos datos.

 **Problema del sobreajuste (Overfitting)**

El **sobreajuste** ocurre cuando un modelo aprende demasiado bien los detalles y el ruido de los datos de entrenamiento, pero no generaliza bien a datos no vistos. Esto sucede cuando el modelo tiene demasiados parámetros o cuando los coeficientes del modelo son demasiado grandes. La regularización ayuda a evitar esto.

 **Tipos de regularización**

Existen varias técnicas de regularización en modelos de regresión, pero las dos más comunes son:

- **Regularización L2 (Ridge Regression)**: Agrega una penalización proporcional al **cuadrado** de los coeficientes del modelo.
- **Regularización L1 (Lasso Regression)**: Agrega una penalización proporcional a la **suma absoluta** de los coeficientes del modelo.

## **Regularización L2 (Ridge Regression)**

En la **regularización L2**, el término de penalización se basa en la **suma de los cuadrados** de los coeficientes del modelo. La fórmula para la regresión Ridge es:

$$
\text{Ridge} \, \text{Loss} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{m} \beta_j^2
$$

Donde:
- $y_i$ es el valor real de la variable dependiente para el i-ésimo dato.
- $\hat{y}_i$ es el valor predicho por el modelo para el i-ésimo dato.
- $n$ es el número de datos de entrenamiento.
- $\beta_j$ son los coeficientes del modelo (para cada característica $j$).
- $\lambda$ es el **parámetro de regularización** que controla la cantidad de penalización sobre los coeficientes.

La penalización $ \lambda \sum_{j=1}^{m} \beta_j^2$ agrega un **castigo** a los coeficientes grandes, lo que obliga al modelo a mantener los coeficientes lo más pequeños posible, ayudando a evitar el sobreajuste.

## **Regularización L1 (Lasso Regression)**

La **regularización L1** funciona de manera similar a L2, pero la penalización se basa en la **suma de los valores absolutos** de los coeficientes. La fórmula para la regresión Lasso es:

$$
\text{Lasso} \, \text{Loss} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{m} |\beta_j|
$$

Donde:
- Los términos son los mismos que en la regularización L2, pero la penalización ahora es sobre el valor absoluto de los coeficientes $|\beta_j|$.
- $\lambda$ es el **parámetro de regularización**.

La regularización L1 tiene una característica única: puede llevar a que algunos coeficientes sean **exactamente cero**, lo que significa que ciertas características pueden ser **eliminadas** del modelo. Este comportamiento lo hace útil cuando se desea **selección de características** (feature selection).

 ## **¿Cómo afecta el parámetro de regularización $\lambda $?**

El parámetro $\lambda$ controla la **intensidad** de la penalización. Si $\lambda = 0$, no se aplica ninguna regularización y el modelo es simplemente una regresión lineal ordinaria. A medida que $\lambda$ aumenta, el modelo penaliza más fuertemente los coeficientes, lo que puede hacer que los coeficientes se vuelvan más pequeños y el modelo más simple. Un valor muy grande de $\lambda$ puede resultar en un modelo que sea **demasiado simple** (subajuste o underfitting).

 **Comparación entre L1 y L2:**

- **Lasso (L1)**:
  - Puede llevar a coeficientes **exactamente cero**.
  - Útil para **selección de características**.
  - Mejora la interpretabilidad del modelo, ya que elimina algunas variables.

- **Ridge (L2)**:
  - No lleva a coeficientes exactamente cero.
  - Funciona bien cuando **todas las características** son relevantes.
  - Es útil cuando tienes muchas variables correlacionadas entre sí.

## **Regularización ElasticNet**

Una combinación de L1 y L2 es el **ElasticNet**, que incluye tanto la penalización L1 como L2. La fórmula para ElasticNet es:

$$
\text{ElasticNet} \, \text{Loss} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda_1 \sum_{j=1}^{m} |\beta_j| + \lambda_2 \sum_{j=1}^{m} \beta_j^2
$$

Donde $\lambda_1$ y $\lambda_2$ controlan la cantidad de regularización L1 y L2, respectivamente. ElasticNet es útil cuando deseas combinar los beneficios de Lasso y Ridge.

 **¿Cuándo usar la regularización?**

La regularización es útil cuando:
- Tienes un **modelo complejo** (con muchos parámetros) y deseas evitar el sobreajuste.
- Estás trabajando con **datos de alta dimensión** (muchas variables) y deseas reducir la complejidad del modelo.
- Quieres realizar **selección de características** o reducir el número de características relevantes.

 **Resumen**

La **regularización** es una técnica fundamental para controlar el sobreajuste en los modelos de regresión. A través de la **penalización** de los coeficientes, la regularización ayuda a simplificar el modelo y a mejorar su capacidad de generalización. Las dos formas más comunes de regularización son:
- **Lasso (L1)**, que puede eliminar características irrelevantes al llevar algunos coeficientes a cero.
- **Ridge (L2)**, que penaliza los coeficientes grandes pero no los lleva a cero.
-  La regularización ayuda a prevenir el sobreajuste al penalizar los coeficientes grandes, lo que puede mejorar la capacidad del modelo para generalizar a nuevos datos. Técnicas como la regularización Lasso (L1) y Ridge (L2) son comunes para este propósito

La elección entre Lasso, Ridge y ElasticNet depende del tipo de problema que estás resolviendo y las características de los datos.



# Función de coste 

La **función de coste** (también conocida como **función de pérdida**) es una métrica que nos permite medir qué tan bien o mal está funcionando un modelo de machine learning. En términos simples, es una **medida de error** entre las predicciones del modelo y los valores reales.

El objetivo de entrenar un modelo de machine learning es **minimizar** esta función de coste para que el modelo haga predicciones lo más cercanas posibles a los valores reales.

 **¿Por qué es importante la función de coste?**

La función de coste es crucial porque:
- **Guía el entrenamiento del modelo**: Durante el entrenamiento, los algoritmos de optimización (como el **gradiente descendente**) utilizan la función de coste para ajustar los parámetros del modelo (como los coeficientes en la regresión) para reducir el error.
- **Evalúa el rendimiento del modelo**: Cuanto menor sea el valor de la función de coste, mejor será el rendimiento del modelo en términos de precisión de las predicciones.

 **Tipos comunes de funciones de coste**

Dependiendo del tipo de problema que estemos resolviendo, se utilizan diferentes funciones de coste. Las más comunes son:

## **Función de coste en regresión**:
  - Para los problemas de regresión, donde se predicen valores continuos, la función de coste más común es el **Error Cuadrático Medio (MSE)**.
  - **MSE** mide la diferencia promedio entre los valores predichos y los reales, elevando al cuadrado las diferencias para penalizar errores grandes.
  
  $$ 
  MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
  $$

  Donde:
  - $y_i$ es el valor real de la variable dependiente para el i-ésimo dato.
  - $\hat{y}_i$ es el valor predicho por el modelo para el i-ésimo dato.
  - $n$ es el número total de datos.

## **Función de coste en clasificación**:
  - Para problemas de clasificación, como clasificación binaria o multiclase, se utiliza una función de coste basada en **entropía cruzada** o **log loss**.
  
  En clasificación binaria, la función de coste **log loss** es:

  $$
  \text{Log Loss} = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]
  $$

  Donde:
  - $y_i$ es el valor real (0 o 1) para el i-ésimo dato.
  - $\hat{y}_i$ es la probabilidad predicha de que el i-ésimo dato pertenezca a la clase 1.
  - $n$ es el número total de datos.

  La entropía cruzada mide la distancia entre las distribuciones de probabilidad de las predicciones del modelo y las verdaderas distribuciones de los datos.

##  **Propósito de la función de coste**

El propósito principal de la función de coste es **guiar el proceso de optimización** para ajustar los parámetros del modelo. A medida que el algoritmo de aprendizaje ajusta los parámetros del modelo (por ejemplo, en la regresión, los coeficientes $\beta$), la función de coste nos dice qué tan bien lo está haciendo, y el objetivo es **minimizarla**. 

El proceso de entrenamiento busca la configuración de parámetros que reduzca el valor de la función de coste al mínimo posible.

 **¿Cómo se minimiza la función de coste?**

En la mayoría de los modelos de machine learning, el **gradiente descendente** es el algoritmo utilizado para minimizar la función de coste. Este algoritmo ajusta los parámetros del modelo en la dirección opuesta al gradiente de la función de coste (es decir, hacia el mínimo).

- **Gradiente Descendente**:
  
  El gradiente descendente es un proceso iterativo que actualiza los parámetros del modelo de la siguiente forma:

  $$
  \theta_j := \theta_j - \alpha \frac{\partial J(\theta)}{\partial \theta_j}
  $$

  Donde:
  - $\theta_j$ es el parámetro del modelo que estamos actualizando.
  - $J(\theta)$ es la función de coste.
  - $\frac{\partial J(\theta)}{\partial \theta_j}$ es el gradiente de la función de coste respecto a $\theta_j$.
  - $\alpha$ es la tasa de aprendizaje, que controla el tamaño de los pasos que damos en la actualización.

 **Tipos de errores y su relación con la función de coste**

La función de coste también nos permite entender la naturaleza de los errores cometidos por el modelo:
- **Sobreajuste (Overfitting)**: Si la función de coste es muy baja en los datos de entrenamiento, pero alta en los datos de prueba, significa que el modelo ha memorizado demasiado los datos de entrenamiento y no está generalizando bien.
- **Subajuste (Underfitting)**: Si la función de coste es alta tanto en los datos de entrenamiento como en los datos de prueba, significa que el modelo no está aprendiendo lo suficiente de los datos y no es lo suficientemente complejo para capturar las relaciones entre las variables.

**Resumen**

- La **función de coste** es una medida de lo bien que un modelo de machine learning está prediciendo los resultados correctos.
- Su objetivo es ser **minimizada** durante el proceso de entrenamiento.
- En **regresión**, las funciones de coste comunes son el **MSE** y el **MAE**.
- En **clasificación**, la **entropía cruzada** o **log loss** es común.
- Los **algoritmos de optimización** como el **gradiente descendente** se utilizan para ajustar los parámetros del modelo y minimizar la función de coste.

La función de coste es fundamental para entender cómo un modelo está aprendiendo y si está **generalizando correctamente** o **ajustándose demasiado** a los datos de entrenamiento.


# Medidas de error útiles para comparar regresiones

Cuando estamos trabajando con modelos de **regresión**, queremos evaluar qué tan bien se ajusta el modelo a los datos. Para ello, usamos diversas **medidas de error** que nos permiten comparar las predicciones del modelo con los valores reales. A continuación, se describen algunas de las medidas más comunes para evaluar y comparar modelos de regresión.

## 1. **Error Cuadrático Medio (MSE - Mean Squared Error)**

El **Error Cuadrático Medio (MSE)** mide la diferencia promedio entre los valores predichos y los valores reales, elevando al cuadrado esas diferencias. **Penaliza más los errores grandes debido a la operación de elevar al cuadrado.**

Fórmula:

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

Donde:
- $y_i$ es el valor real.
- $\hat{y}_i$ es el valor predicho por el modelo.
- $n$ es el número total de datos.

**Ventajas**:
- Penaliza más los errores grandes (por lo que ayuda a prevenir modelos con grandes errores).
- Es fácil de calcular y comprender.

**Desventajas**:
- Al estar en unidades cuadradas, es difícil de interpretar en términos de las unidades de los datos originales.

## 2. **Raíz del Error Cuadrático Medio (RMSE - Root Mean Squared Error)**

El **RMSE** es simplemente la raíz cuadrada del **MSE**. Esta medida es útil porque devuelve el error a las unidades originales de los datos, lo que la hace más fácil de interpretar.

Fórmula:

$$
RMSE = \sqrt{MSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}
$$

**Ventajas**:
- Está en las mismas unidades que los datos, lo que facilita su interpretación.
- Penaliza de forma más significativa los errores grandes, al igual que el MSE.

**Desventajas**:
- Al igual que el MSE, es sensible a los **outliers** (valores atípicos) debido a la operación de elevar al cuadrado.

## 3. **Error Absoluto Medio (MAE - Mean Absolute Error)**

El **Error Absoluto Medio (MAE)** mide la diferencia promedio entre los valores reales y los valores predichos, pero sin elevar al cuadrado las diferencias, lo que lo hace menos sensible a los errores grandes en comparación con el MSE o RMSE.

Fórmula:

$$
MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

**Ventajas**:
- Es fácil de interpretar, ya que está en las mismas unidades que los datos.
- No penaliza los errores grandes de manera tan fuerte como el MSE y el RMSE.

**Desventajas**:
- No distingue entre grandes y pequeños errores de la misma manera que el MSE o RMSE. No da un castigo fuerte por errores grandes.

## 4. **Error Absoluto Mediano (MedAE - Median Absolute Error)**

El **Error Absoluto Mediano (MedAE)** es similar al MAE, pero en lugar de tomar el promedio de las diferencias absolutas, toma la **mediana** de esos valores. Esto lo hace más robusto frente a **outliers**.

Fórmula:

$$
MedAE = \text{mediana} \left( |y_i - \hat{y}_i| \right)
$$

**Ventajas**:
- Muy robusto frente a **valores atípicos**.
- Puede ser útil cuando los datos contienen muchos outliers que no se deben penalizar excesivamente.
**Desventajas**:
- Menos sensible a la magnitud de los errores grandes que el MSE y RMSE.

## 5. **Coeficiente de Determinación ($R^2$)**

El **coeficiente de determinación ($R^2$)** mide la proporción de la variabilidad total de los datos que es explicada por el modelo. Se utiliza para determinar qué tan bien se ajusta el modelo a los datos.

Fórmula:

$$
R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}
$$

Donde:
- $y_i$ son los valores reales.
- $\hat{y}_i$ son los valores predichos por el modelo.
- $\bar{y}$ es la media de los valores reales.

**Ventajas**:
- Proporciona una idea de la proporción de la varianza de los datos que el modelo está explicando.
- Si $R^2$ es cercano a 1, significa que el modelo explica bien los datos.

**Desventajas**:
- Puede ser **engañoso** cuando se usa en modelos que no tienen una relación lineal con los datos.
- No siempre indica un buen modelo si hay una gran variabilidad en los datos o si el modelo tiene muchas variables.

## 6. **Error Porcentual Absoluto Medio (MAPE - Mean Absolute Percentage Error)**

El **MAPE** mide el error en términos porcentuales, es decir, qué tan lejos están las predicciones en relación con los valores reales, en porcentaje.

Fórmula:

$$
MAPE = \frac{100}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right|
$$

**Ventajas**:
- Es fácil de interpretar, ya que está en porcentaje.
- Se puede comparar el rendimiento entre diferentes conjuntos de datos de manera intuitiva.

**Desventajas**:
- Puede ser **problemático** si los valores reales $y_i$ son cercanos a cero, ya que esto provocaría valores de error muy altos o indefinidos.

## 7. **Errores Simples vs. Ponderados**

Es importante notar que también puedes ponderar ciertos errores si crees que algunos datos o características son más importantes que otros. Por ejemplo, en un conjunto de datos donde algunos valores son más críticos para la predicción, puedes usar una **función de coste ponderada**.

##  **Resumen de las principales métricas**

| Métrica              | Fórmula                                  | Pros                                          | Contras                                      |
|----------------------|------------------------------------------|-----------------------------------------------|----------------------------------------------|
| **MSE**              | $\frac{1}{n} \sum (y_i - \hat{y}_i)^2$   | Penaliza errores grandes. Fácil de calcular.  | Sensible a outliers.                         |
| **RMSE**             | $\sqrt{MSE}$                             | En las mismas unidades que los datos.         | Sensible a outliers.                         |
| **MAE**              | $\frac{1}{n} \sum |y_i - \hat{y}_i|$    | Fácil de interpretar. No penaliza errores grandes. | No distingue bien entre errores grandes y pequeños. |
| **MedAE**            | Mediana de $|y_i - \hat{y}_i|$           | Robusto frente a outliers.                    | Menos sensible a errores grandes.            |
| **$R^2$**            | $1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$ | Mide el ajuste global del modelo.              | No siempre es confiable, especialmente en modelos no lineales. |
| **MAPE**             | $\frac{100}{n} \sum \left| \frac{y_i - \hat{y}_i}{y_i} \right|$ | En términos porcentuales, fácil de interpretar. | Problemático con valores cercanos a cero.    |

 **¿Cuál debo usar?**
La elección de la medida de error depende del tipo de modelo y el contexto. Aquí algunos consejos:
- Si te importa **penalizar los errores grandes**, usa **MSE** o **RMSE**.
- Si prefieres una métrica que sea más **robusta a outliers**, utiliza **MAE** o **MedAE**.
- Si necesitas entender **qué porcentaje del error es relevante**, el **MAPE** es útil.
- Si quieres saber **cuánto varían los datos que tu modelo puede explicar**, usa **$R^2$**.
