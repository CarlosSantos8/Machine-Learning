# Cross Validation Approaches

La **validación cruzada** (cross-validation) es una técnica utilizada para evaluar el rendimiento de un modelo de aprendizaje automático, dividiendo el conjunto de datos en varias partes y evaluando el modelo en diferentes subconjuntos. Esto ayuda a obtener una mejor estimación de cómo se comportará el modelo en datos no vistos. A continuación, se describen tres enfoques comunes de validación cruzada: **K-fold Cross Validation**, **Leave One Out Cross Validation (LOOCV)** y **Stratified Cross Validation**.


La validación cruzada es una técnica estadística empleada en el aprendizaje automático para evaluar el rendimiento y la generalización de un modelo predictivo. Su objetivo principal es dividir un conjunto de datos en subconjuntos, entrenar el modelo con algunos de ellos y evaluar su rendimiento con el subconjunto restante. Repetimos este proceso varias veces y promediamos las métricas de rendimiento del modelo a lo largo de estas iteraciones.

La validación cruzada ofrece varias ventajas, entre ellas una evaluación del rendimiento más fiable al reducir el impacto de la variabilidad en una única división aleatoria de datos. Además, optimiza el uso de los datos disponibles, garantizando que cada punto de datos se utilice para la validación solo una vez.
La forma más común de validación cruzada es la validación cruzada de k-fold .

En la validación cruzada de k pliegues, el conjunto de datos se divide aleatoriamente en k pliegues o subconjuntos de igual tamaño.

Posteriormente, el modelo se entrena con k-1 pliegues y se valida con el pliegue restante. Este ciclo se itera k veces, y cada iteración utiliza un pliegue único como conjunto de prueba.
La métrica de rendimiento final suele ser el promedio de las métricas obtenidas en cada iteración.

Por ejemplo, si elige k = 5, creará cinco pliegues. Con este enfoque, entrenará su modelo cinco veces. En cada iteración, un pliegue diferente constituye el conjunto de validación, y entrenará su modelo con los otros cuatro pliegues.
En consecuencia, cada punto de datos puede formar parte del conjunto de validación solo una vez.

# Tipos de validación Cruzada

##  K-fold Cross Validation

La **K-fold Cross Validation** es uno de los métodos más populares de validación cruzada. Este método divide el conjunto de datos en **K** partes o **folds** de tamaño similar.

En la validación cruzada tradicional de K-Fold, el conjunto de datos se particiona aleatoriamente en K pliegues de tamaño aproximadamente igual. En cada iteración, un pliegue constituye el conjunto de validación, mientras que los K-1 pliegues restantes constituyen el conjunto de entrenamiento. Repetimos este procedimiento K veces, asegurándonos de que cada pliegue se utilice solo una vez como conjunto de validación.

**Funcionamiento:**
- El conjunto de datos se divide en **K** pliegues (subconjuntos).
- El modelo se entrena **K** veces, utilizando **K-1** pliegues como datos de entrenamiento y el pliegue restante como conjunto de prueba.
- Este proceso se repite **K** veces, cada vez con un pliegue diferente como conjunto de prueba, y el rendimiento del modelo se evalúa promediando los resultados obtenidos en cada una de las iteraciones.

**Ventajas:**
- Utiliza todos los datos para entrenar y probar el modelo.
- Ayuda a evitar el sobreajuste (overfitting) al evaluar el modelo en múltiples conjuntos de prueba.

### Desventajas:
- Es más costoso computacionalmente, ya que el modelo debe ser entrenado varias veces (una vez por cada pliegue).
- Los resultados pueden variar según el valor de **K**.


## 2. Leave One Out Cross Validation (LOOCV)

**Leave One Out Cross Validation (LOOCV)** es un caso especial de la **K-fold Cross Validation**, donde **K** es igual al número total de muestras en el conjunto de datos. En otras palabras, en cada iteración, se utiliza solo una muestra como conjunto de prueba y el resto como conjunto de entrenamiento.

### Funcionamiento:
- En cada iteración, el modelo se entrena utilizando todos los datos, excepto una sola muestra, que se utiliza para evaluar el modelo.
- Este proceso se repite para cada muestra del conjunto de datos, lo que significa que el modelo se entrena tantas veces como muestras hay en el conjunto de datos.

### Ventajas:
- Cada muestra del conjunto de datos se utiliza tanto para entrenar como para evaluar el modelo, lo que puede ser útil especialmente con conjuntos de datos pequeños.
- No pierde ningún dato, ya que cada observación se usa para evaluación.

### Desventajas:
- Es muy costoso en términos computacionales, especialmente cuando el conjunto de datos es grande, ya que el modelo debe entrenarse tantas veces como muestras haya.
- Puede ser muy sensible a puntos de datos atípicos (outliers), ya que cada evaluación depende de una única muestra.


## 3. Stratified Cross Validation

La **Stratified Cross Validation** es una variante de la **K-fold Cross Validation** que asegura que cada pliegue tenga la misma proporción de clases que el conjunto de datos original. Este enfoque es especialmente útil cuando se trabaja con conjuntos de datos desbalanceados, donde algunas clases pueden tener muchas más muestras que otras.

Si bien la selección de datos no es totalmente aleatoria, existe un elemento de aleatoriedad en la elección de muestras dentro de cada clase.

Por ejemplo, considere un escenario de clasificación binaria con un conjunto de datos desequilibrado, donde el 90 % de las muestras pertenecen a la clase A y el 10 % a la clase B. En una validación cruzada estratificada de 5 pliegues, cada pliegue incluirá una representación proporcional de muestras de clase A y clase B , preservando de manera consistente la relación de desequilibrio entre los pliegues.


### Funcionamiento:
- En lugar de dividir aleatoriamente el conjunto de datos, se realiza una división en la que se mantiene la misma distribución de clases en cada pliegue. Esto asegura que el modelo se entrene y evalúe de manera más representativa respecto a las proporciones de cada clase.
- Este enfoque puede ser implementado tanto en **K-fold Cross Validation** como en **LOOCV**, garantizando que cada pliegue contenga una representación equilibrada de las clases.

### Ventajas:
- Asegura que cada pliegue sea representativo de la distribución de clases, lo que es especialmente importante cuando se tiene un conjunto de datos desbalanceado.
- Mejora la capacidad de generalización del modelo al evitar que el modelo se entrene con datos que no representen adecuadamente las clases.

### Desventajas:
- A pesar de mejorar la representación de las clases, puede seguir siendo menos eficiente si las clases están extremadamente desbalanceadas o si el tamaño de la muestra es pequeño.
- Puede ser más costoso computacionalmente que la validación cruzada tradicional si se requiere una partición más compleja.

## Comparación de Métodos

| Método                        | Ventajas                                                                 | Desventajas                                                           |
|-------------------------------|--------------------------------------------------------------------------|-----------------------------------------------------------------------|
| **K-fold Cross Validation**    | Utiliza todo el conjunto de datos para entrenamiento y evaluación.       | Más costoso computacionalmente a medida que **K** aumenta.            |
| **Leave One Out Cross Validation (LOOCV)** | Utiliza todos los datos para entrenar el modelo, ideal para conjuntos de datos pequeños. | Muy costoso computacionalmente.                                      |
| **Stratified Cross Validation**| Asegura una división equilibrada de las clases, ideal para datos desbalanceados. | Más complejo de implementar, especialmente con grandes conjuntos de datos. |



Cada una de estas técnicas de validación cruzada tiene sus ventajas y desventajas, y la elección de cuál usar depende del problema específico, el tamaño del conjunto de datos y los recursos computacionales disponibles. En general, la **K-fold Cross Validation** es un buen punto de partida, mientras que **LOOCV** es útil para conjuntos de datos pequeños. La **Stratified Cross Validation** es esencial cuando se trabaja con conjuntos de datos desbalanceados, especialmente en problemas de clasificación.

- Los parámetros del modelo se determinarán según nuestros datos de entrenamiento.
- Lo llamamos **validación** porque los datos de prueba generalmente se consideran datos que no analizaremos hasta que hayamos terminado de modelar por completo. Mientras que aquí estamos incorporando estos datos de validación e intentando averiguar cual es el mejor modelo.


## Relación entre las métricas de error y la complejidad del modelo

Ahora veremos cómo se relacionan las métricas de error con la **complejidad** cuando estamos determinando nuestro modelo.

- A medida que la **complejidad** crece, la tasa de error **disminuye** .

Cuando realices la **validación cruzada** y obtengas los diferentes conjuntos de resistencia, el objetivo es reducir el error al aumentar la complejidad del modelo. Sin embargo, **llegará un punto de inflexión** donde, al seguir aumentando la complejidad, el modelo empezará a **adaptarse excesivamente** a los datos de entrenamiento, lo que provocará que no se generalice bien a nuevos datos.

### Modelos con baja complejidad

Son modelos relativamente simples, entrenados y construidos con un número limitado de características. Son fáciles de entender e interpretar, y funcionan bien con conjuntos de datos pequeños.
Estos modelos no pueden capturar relaciones complejas ni son flexibles para adaptarse a diversos patrones de datos.
Son propensos al subajuste, lo que significa que no proporcionan una buena precisión en los datos de entrenamiento y prueba.

Los modelos asociados con el **lado izquierdo de la curva** se consideran **inadecuados**. Esto se debe a que no son lo suficientemente complejos y, como resultado:
- El error de entrenamiento y de validación cruzada es muy alto, lo que lleva a un **underfitting** (ajuste insuficiente).

> **Nota:** El objetivo es encontrar un equilibrio en la complejidad para que el modelo no caiga en el **underfitting** ni en el **overfitting**.


![Model_ComplexityvsError_Underfitting](.\images\Model_ComplexityvsError_Underfitting.png)


### Modelos con alta complejidad

Son modelos sofisticados que se construyen y entrenan con más características. Son más flexibles al adaptarse a diversos patrones de datos. Esta flexibilidad también puede causar sobreajuste, donde el modelo puede memorizar ruido o patrones específicos que no se generalizan bien al probar los datos.

Los modelos asociados con el **lado derecho de la curva** se consideran **sobreajustados** a los datos, es decir, se ajustan demasiado a los datos de entrenamiento y no generalizan bien para datos nuevos.

- El **error de validación cruzada** aumenta a medida que aumenta la complejidad del modelo. Esto ocurre porque el modelo comienza a capturar el **ruido** o las variaciones aleatorias en los datos de entrenamiento, en lugar de aprender patrones generales aplicables a nuevos datos.



![Model_ComplexityvsError_over](.\images\Model_ComplexityvsError_over.jpg)

### El "Goldilocks Point": El punto óptimo de complejidad

El **"Goldilocks Point"** (punto de oro) es el momento clave en el que el modelo tiene la **complejidad justa** para generalizar bien. Este es el punto donde el modelo no es ni demasiado simple ni demasiado complejo.

- **Antes del Goldilocks Point**: El modelo es demasiado simple (underfitting), lo que significa que no está capturando los patrones suficientes en los datos.
- **Después del Goldilocks Point**: El modelo se vuelve demasiado complejo (overfitting), lo que significa que está ajustando demasiado los datos de entrenamiento y pierde su capacidad de generalización.

En este punto óptimo, el **error de validación cruzada** es el **más bajo** y el modelo tiene el **mejor desempeño en datos no vistos**. Encontrar este punto es esencial para entrenar modelos que no sufran de underfitting ni de overfitting.

# Diferencias entre `cross_val_score`, `KFold` y `StratifiedKFold`


![cross_validation](.\images\cross_validation.jpg)

## 1. `cross_val_score`

La función **`cross_val_score`** es una forma rápida y sencilla de realizar validación cruzada en un modelo de Machine Learning. Esta función divide automáticamente los datos en varias particiones, entrena el modelo en cada partición y evalúa su rendimiento. Devuelve las puntuaciones obtenidas en cada iteración de la validación cruzada.

### Parámetros:
- **`estimator`**: El modelo o estimador que será evaluado (por ejemplo, un clasificador o regresor).
- **`X`**: Datos de entrada (features).
- **`y`**: Etiquetas de salida (target).
- **`cv`**: Número de particiones (folds) o un generador de particiones como `KFold` o `StratifiedKFold`.
- **`scoring`**: Métrica que se usará para evaluar el rendimiento (por ejemplo, "accuracy", "f1", etc.).

## 2. `KFold`

**`KFold`** es una técnica de validación cruzada que se utiliza para evaluar el rendimiento de un modelo de Machine Learning. Esta técnica divide el conjunto de datos en **k** particiones o *folds*, de manera que en cada iteración se entrena el modelo en **k-1** particiones y se evalúa en la partición restante. Este proceso se repite **k** veces, utilizando cada partición como conjunto de prueba una vez.

La validación cruzada con **`KFold`** permite tener una evaluación más robusta del modelo, ya que se aprovechan todas las muestras tanto para entrenamiento como para validación.

**¿Cómo Funciona?**

1. **División de los datos**: Los datos se dividen en **k** particiones de manera aleatoria.
2. **Entrenamiento y validación**: En cada iteración, se entrena el modelo con **k-1** particiones y se evalúa con la partición restante.
3. **Repetición**: Este proceso se repite **k** veces, de forma que cada partición es usada una vez como conjunto de prueba.
4. **Promediado de resultados**: Los resultados de las **k** iteraciones se promedian para obtener una estimación más precisa del rendimiento del modelo.

**Parámetros de `KFold`**

- **`n_splits`**: Número de particiones (folds) en que se dividirán los datos.
- **`shuffle`**: Si se debe aleatorizar el orden de los datos antes de realizar la división. El valor predeterminado es `False`, lo que significa que los datos se dividen de manera secuencial.
- **`random_state`**: Semilla para la aleatorización, que asegura que el proceso sea reproducible. Si no se establece, se obtiene una aleatorización diferente en cada ejecución.
- **`test_size`**: Proporción de los datos que se utilizarán para validación en cada fold. Este parámetro no se usa directamente en `KFold`, pero se puede controlar mediante el número de splits.


## 3. `StratifiedKFold`

**`StratifiedKFold`** es una variante de **`KFold`** que garantiza que cada partición (fold) tenga una distribución de clases similar a la del conjunto original. Esto es especialmente útil cuando se trabaja con conjuntos de datos desbalanceados, es decir, cuando algunas clases están subrepresentadas en comparación con otras.

En lugar de dividir los datos de manera completamente aleatoria, **`StratifiedKFold`** realiza la partición de forma que se mantenga la proporción de las clases en cada uno de los folds, asegurando que cada partición tenga una representación representativa de las clases.

**¿Por qué Usar `StratifiedKFold`?**

En problemas de clasificación con clases desbalanceadas, el uso de **`KFold`** podría causar que algunas particiones no tengan suficientes muestras de las clases minoritarias. Esto puede afectar negativamente la evaluación del modelo, ya que algunos folds podrían no contener ejemplos de las clases menos frecuentes.

**`StratifiedKFold`** soluciona este problema manteniendo la distribución de clases en cada fold, lo que permite una evaluación más precisa del modelo.

**¿Cómo Funciona?**

1. **División de los datos**: Al igual que **`KFold`**, los datos se dividen en **k** particiones (folds). Sin embargo, en lugar de dividirlos aleatoriamente, **`StratifiedKFold`** asegura que la distribución de clases en cada partición sea proporcional a la distribución global del conjunto de datos.
2. **Entrenamiento y validación**: Al igual que en **`KFold`**, el modelo se entrena con **k-1** particiones y se evalúa con la partición restante.
3. **Repetición**: Este proceso se repite **k** veces, usando cada partición como conjunto de prueba una vez.
 
**Parámetros de `StratifiedKFold`**

- **`n_splits`**: Número de particiones (folds) en que se dividirán los datos.
- **`shuffle`**: Si se debe aleatorizar el orden de los datos antes de realizar la división. El valor predeterminado es `False`.
- **`random_state`**: Semilla para la aleatorización, que asegura que el proceso sea reproducible.
- **`stratify`**: Este parámetro asegura que la proporción de clases se mantenga en cada partición.




**Un valor k menor (k = 2 o 3)**

Resulta en una mayor varianza entre pliegues, ya que cada iteración se entrena con una porción menor del conjunto de datos. Dado que los datos de entrenamiento son más limitados en cada iteración, el rendimiento del modelo fluctúa más según el subconjunto específico de datos utilizado para el entrenamiento y la validación. Esta mayor varianza puede generar estimaciones del modelo menos estables y un mayor riesgo de sobreajuste a conjuntos de entrenamiento más pequeños.

**Un k mayor (k = 10 o 20)**
Resulta en una menor varianza entre los pliegues, ya que cada iteración se entrena con una porción mayor del conjunto de datos. Dado que el conjunto de entrenamiento es casi el conjunto de datos completo en cada iteración, el rendimiento del modelo es más estable en diferentes pliegues. Esto ayuda a mejorar la generalización y reduce el sobreajuste, ya que el modelo aprende de un conjunto de datos más diverso en cada iteración.

Una menor varianza significa que el modelo funciona de forma más consistente en todos los pliegues, con fluctuaciones mínimas en métricas de rendimiento como el error cuadrático medio. Sin embargo, **el coste computacional aumenta a medida que aumenta k**, ya que el modelo debe entrenarse y evaluarse k veces.

**Ejemplo:** Consideremos un escenario real en el que construimos un modelo de regresión para predecir los precios de las viviendas en función de factores como la superficie, el número de habitaciones y el barrio. Al usar k = 10, el modelo se entrena con el 90 % de los datos y prueba con el 10 % en cada iteración. Dado que cada conjunto de entrenamiento es amplio y diverso, el modelo se generaliza correctamente a datos no analizados.

- Pliegue 1: Entrenamos el modelo en un subconjunto donde predominan las casas con gran superficie y muchos dormitorios. El modelo aprendido puede priorizar estas características como cruciales para predecir los precios de las viviendas.

- Pliegue 2: A continuación, entrenamos el modelo en un subconjunto diferente donde las casas en un barrio diverso, independientemente de su superficie, tienden a tener precios más altos. En este caso, el modelo aprendido asigna mayor importancia al barrio que en el Pliegue 1.

Por lo tanto, estos dos subconjuntos muestran una variación considerable.
